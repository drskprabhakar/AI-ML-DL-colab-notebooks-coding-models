Classification Rules (Using Steep Meridian)

Type	Steep meridian axis (degrees)

WTR	90 ± 30 → 60–120° (vertical steep meridian)

ATR	0 ± 30 → 0–30° or 150–180° (horizontal steep meridian)

Oblique	All others (31–59° and 121–149°)

## Note: Since PreopA1 is flatter meridian, we add 90° and take modulo 180 to get the steep meridian axis.

```python
# Function to classify astigmatism
def classify_astig(r1_axis):
    steep_axis = (r1_axis + 90) % 180
    if 60 <= steep_axis <= 120:
        return 'WTR'
    elif (0 <= steep_axis <= 30) or (150 <= steep_axis <= 180):
        return 'ATR'
    else:
        return 'Oblique'

# Apply classification
sorted_data['Preop_Astig_Type'] = sorted_data['PreopA1'].apply(classify_astig)

# Count each type
type_counts = sorted_data['Preop_Astig_Type'].value_counts()
type_counts
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Create cross-tabulation
cross_tab = pd.crosstab(sorted_data['Preop_Astig_Type'], sorted_data['Cornea_Type'])

# Calculate percentages within each astigmatism type
cross_tab_perc = cross_tab.div(cross_tab.sum(axis=1), axis=0) * 100

# Colors for cornea types
colors = ['lightblue','blue','dodgerblue']

# Plot grouped bar chart
ax = cross_tab.plot(kind='bar', figsize=(10,6), color=colors)

plt.title('Astigmatism Type vs Cornea Type (%)', fontsize=14)
plt.xlabel('Astigmatism Type', fontsize=12)
plt.ylabel('Number of Eyes', fontsize=12)
plt.xticks(rotation=0)
plt.legend(title='Cornea Type')

# Add percentage annotations
for i, ast_type in enumerate(cross_tab.index):
    for j, cor_type in enumerate(cross_tab.columns):
        count = cross_tab.loc[ast_type, cor_type]
        perc = cross_tab_perc.loc[ast_type, cor_type]
        ax.text(i + j*0.25 - 0.25, count + 0.3, f'{perc:.1f}%', ha='center', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# Cylinder and axis
C = sorted_data['preop_astigmatism']   # magnitude in diopters
A_deg = sorted_data['PreopA1']         # axis in degrees

# Convert axis to radians
A_rad = np.deg2rad(A_deg)

# Calculate J0 and J45
sorted_data['J0'] = - (C / 2) * np.cos(2 * A_rad)
sorted_data['J45'] = - (C / 2) * np.sin(2 * A_rad)

# Show first few rows
sorted_data[['preop_astigmatism','PreopA1','J0','J45']].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Calculate centroid magnitude and axis
centroids = sorted_data.groupby('Preop_Astig_Type')[['J0','J45']].mean().reset_index()
centroids['Centroid_Mag'] = 2 * np.sqrt(centroids['J0']**2 + centroids['J45']**2)
centroids['Centroid_Axis'] = 0.5 * np.degrees(np.arctan2(centroids['J45'], centroids['J0']))

# Colors for astigmatism types
colors = {'WTR':'skyblue', 'ATR':'orange', 'Oblique':'lightgreen'}

# Set style
sns.set(style="whitegrid")

# Scatter plot of individual vectors
plt.figure(figsize=(8,8))
sns.scatterplot(x='J0', y='J45', hue='Preop_Astig_Type', data=sorted_data,
                s=60, palette=colors, edgecolor='k', alpha=0.6)

# Overlay centroids
sns.scatterplot(x='J0', y='J45', hue='Preop_Astig_Type', data=centroids,
                s=250, palette=colors, marker='X', edgecolor='k', legend=False)

# Draw arrows and annotate magnitude & axis
for i, row in centroids.iterrows():
    plt.arrow(0, 0, row['J0'], row['J45'], color=colors[row['Preop_Astig_Type']],
              width=0.005, head_width=0.03, head_length=0.03, length_includes_head=True)
    plt.text(row['J0']*1.05, row['J45']*1.05,
             f"{row['Centroid_Mag']:.2f} D\n{row['Centroid_Axis']:.1f}°",
             color=colors[row['Preop_Astig_Type']], fontsize=10, fontweight='bold')

# Add horizontal and vertical reference lines at 0
plt.axhline(0, color='red', linestyle='--', linewidth=1)
plt.axvline(0, color='red', linestyle='--', linewidth=1)

plt.title('Preoperative Astigmatism Vectors with Centroids, Magnitude & Axis', fontsize=14)
plt.xlabel('J0 (D)', fontsize=12)
plt.ylabel('J45 (D)', fontsize=12)
plt.xlim(sorted_data['J0'].min() - 0.5, sorted_data['J0'].max() + 0.5)
plt.ylim(sorted_data['J45'].min() - 0.5, sorted_data['J45'].max() + 0.5)
plt.legend(title='Astigmatism Type')
plt.grid(True)
plt.show()
```
