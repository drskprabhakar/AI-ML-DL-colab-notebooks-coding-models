Classification Rules (Using Steep Meridian)

Type	Steep meridian axis (degrees)

WTR	90 ± 30 → 60–120° (vertical steep meridian)

ATR	0 ± 30 → 0–30° or 150–180° (horizontal steep meridian)

Oblique	All others (31–59° and 121–149°)

## Note: Since PreopA1 is flatter meridian, we add 90° and take modulo 180 to get the steep meridian axis.

```python
# Function to classify astigmatism
def classify_astig(r1_axis):
    steep_axis = (r1_axis + 90) % 180
    if 60 <= steep_axis <= 120:
        return 'WTR'
    elif (0 <= steep_axis <= 30) or (150 <= steep_axis <= 180):
        return 'ATR'
    else:
        return 'Oblique'

# Apply classification
sorted_data['Preop_Astig_Type'] = sorted_data['PreopA1'].apply(classify_astig)

# Count each type
type_counts = sorted_data['Preop_Astig_Type'].value_counts()
type_counts
```

Standard clinical classification of postcataract refractive errors

✅ Simple myopic astigmatism (SMA)

✅ Compound myopic astigmatism (CMA)

✅ Simple hyperopic astigmatism (SHA)

✅ Compound hyperopic astigmatism (CHA)

✅ Mixed astigmatism (MA)

Classification is based on the sign of the meridional refractions, which come from sphere + cylinder

Step 1 — Compute the two principal meridians

For each eye:

M1 (most plus meridian) = Sphere

M2 (most minus meridian) = Sphere + Cylinder

Step 2

Classification	                 M1	  M2	      Interpretation


Simple Myopic Astigmatism (SMA)	plano	myopic	One meridian 0, the other myopic

Simple Hyperopic Astigmatism (SHA)	plano	hyperopic	One meridian 0, the other hyperopic

Compound Myopic Astigmatism (CMA)	myopic	myopic	Both meridians myopic

Compound Hyperopic Astigmatism (CHA)	hyperopic	hyperopic	Both meridians hyperopic

Mixed Astigmatism (MA)	hyperopic	myopic	One meridian hyperopic, one myopic


```python
def classify_astigmatism(M1, M2):
    if M1 < 0 and M2 < 0:
        return "Compound Myopic Astigmatism"
    elif M1 > 0 and M2 > 0:
        return "Compound Hyperopic Astigmatism"
    elif (M1 == 0 and M2 < 0) or (M2 == 0 and M1 < 0):
        return "Simple Myopic Astigmatism"
    elif (M1 == 0 and M2 > 0) or (M2 == 0 and M1 > 0):
        return "Simple Hyperopic Astigmatism"
    elif (M1 > 0 and M2 < 0) or (M1 < 0 and M2 > 0):
        return "Mixed Astigmatism"
    else:
        return "Other / Emmetropic"
```

```python
def check_cylinder_format(df, cyl_columns):
    """
    Detect if cylinder is mostly minus or plus format.
    Returns a dictionary with the format for each column.
    """
    format_dict = {}
    for col in cyl_columns:
        n_positive = (df[col] > 0).sum()
        n_negative = (df[col] < 0).sum()
        if n_negative >= n_positive:
            format_dict[col] = 'minus'
        else:
            format_dict[col] = 'plus'
    return format_dict
```

```python
def convert_to_minus(df, cyl_columns, sph_columns):
    """
    Converts cylinder in plus format to minus format.
    Updates both cylinder and sphere columns accordingly.
    Formula:
      - Cyl_minus = -Cyl_plus
      - Sph_minus = Sph_plus + Cyl_plus
    """
    format_dict = check_cylinder_format(df, cyl_columns)

    for cyl_col, sph_col in zip(cyl_columns, sph_columns):
        if format_dict[cyl_col] == 'plus':
            df[cyl_col] = -df[cyl_col]
            df[sph_col] = df[sph_col] + (-df[cyl_col])
            print(f"Converted {cyl_col} from plus to minus format.")
        else:
            print(f"{cyl_col} already in minus format.")

    return df
```

```python
def classify_astigmatism(cyl, sph):
    """
    cyl in minus format
    sph = spherical equivalent
    Returns a string label for astigmatism type
    """
    if cyl == 0:
        return 'Simple'
    if cyl > 0:
        if sph * cyl > 0:
            return 'Compound Hyperopic'
        else:
            return 'Mixed'
    elif cyl < 0:
        if sph * cyl > 0:
            return 'Compound Myopic'
        else:
            return 'Mixed'
```
