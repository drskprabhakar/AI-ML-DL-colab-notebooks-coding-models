````python
#m1 = pre-op KV - pre-op KV
df['m1'] = df['pre-op KV']-df['pre-op KH']
df
````

````python
#m1 = pre-op KV - pre-op KV
df['m2'] = df['post-op KV']-df['post-op KH']
df
````

````python
df['a1'] = df['pre-op Axis']
df['a2']= df['post-op Axis']
df
````

````python
df['x1'] = df['m1']*np.cos(df['a1']*2)
df['y1'] = df['m1']*np.sin(df['a1']*2)
df['x2'] = df['m2']*np.cos(df['a2']*2)
df['y2'] = df['m2']*np.sin(df['a2']*2)
df
````

````python
df['Xsia']=df['x2']-df['x1']
df['Ysia']=df['y2']-df['y1']
df
````

````python
# फाइंड sia magnitude
df['SIA-magnitude'] = np.sqrt(df['Xsia']**2+df['Ysia']**2)
df
````

````python
# find sia axis
df['SIA-axis(radian)'] = np.arctan2(df['Ysia'], df['Xsia'])
df
````

````python
# convert radian to degree
df['SIA-axis(degree)'] = df['SIA-axis(radian)']*180/np.pi
df
````

````python
# prompt: Quadrantic handling

# Convert SIA axis to 0-180 degree range
df['SIA-axis(degree) adjusted'] = df['SIA-axis(degree)'].apply(lambda x: x + 180 if x < 0 else x)
df['SIA-axis(degree) adjusted'] = df['SIA-axis(degree) adjusted'].apply(lambda x: x - 180 if x >= 180 else x)

print(df[['SIA-axis(degree)', 'SIA-axis(degree) adjusted']])
df
````

````python
# prompt: Double angle vector plot

import matplotlib.pyplot as plt
import numpy as np
# Function to plot the SIA vector for a given patient index
def plot_sia_vector(patient_index, df):
    sia_x = df.loc[patient_index, 'Xsia']
    sia_y = df.loc[patient_index, 'Ysia']

    # Get the Patient ID if the column exists, otherwise use the index
    if 'Patient ID' in df.columns:
      patient_identifier = df.loc[patient_index, 'Patient ID']
      title_identifier = f"Patient ID: {patient_identifier}"
    else:
      patient_identifier = patient_index
      title_identifier = f"Patient Index: {patient_identifier}"


    # Create the vector plot
    plt.figure(figsize=(6, 6))
    # Draw the original vector from (0,0)
    plt.quiver(0, 0, sia_x, sia_y, angles='xy', scale_units='xy', scale=1, color='red', label='SIA Vector')

    # Draw a second vector representing the angle from (0,0)
    # We can use the magnitude and axis for this
    magnitude = df.loc[patient_index, 'SIA-magnitude']
    axis_rad = df.loc[patient_index, 'SIA-axis(radian)']

    # Draw a smaller vector aligned with the angle
    angle_vector_length = magnitude * 0.5 # You can adjust this length
    angle_vec_x = angle_vector_length * np.cos(axis_rad)
    angle_vec_y = angle_vector_length * np.sin(axis_rad)

    plt.quiver(0, 0, angle_vec_x, angle_vec_y, angles='xy', scale_units='xy', scale=1, color='blue', label='Angle Indicator')


    # Set limits and labels
    # Adjust limits based on the maximum SIA magnitude to ensure all vectors are visible
    max_sia_magnitude = df['SIA-magnitude'].max()
    limit = max(abs(sia_x), abs(sia_y), max_sia_magnitude) * 1.2 # Add some padding
    plt.xlim(-limit, limit)
    plt.ylim(-limit, limit)
    plt.xlabel('SIA X Component')
    plt.ylabel('SIA Y Component')
    plt.title(f'Double Angle Vector Plot for {title_identifier}')
    plt.grid(True)
    plt.axhline(0, color='grey', lw=0.5)
    plt.axvline(0, color='grey', lw=0.5)
    plt.gca().set_aspect('equal', adjustable='box')

    # Add text label for magnitude and axis
    axis_deg = df.loc[patient_index, 'SIA-axis(degree) adjusted']
    # Adjust text position slightly if needed
    text_offset_x = 0.05 * limit * (1 if sia_x >= 0 else -1) # Offset based on vector direction
    text_offset_y = 0.05 * limit * (1 if sia_y >= 0 else -1) # Offset based on vector direction

    # Adjust text alignment based on which quadrant the vector points to
    ha = 'left' if sia_x >= 0 else 'right'
    va = 'bottom' if sia_y >= 0 else 'top'

    plt.text(sia_x, sia_y, f' Mag: {magnitude:.2f}\n Axis: {axis_deg:.2f}°', fontsize=9, ha=ha, va=va)

    plt.legend()
    plt.show()

# Example usage: Plot for the first patient (index 0)
plot_sia_vector(0, df)

# Example usage: Plot for the second patient (index 1)
plot_sia_vector(1, df)

# You can loop through all patients if needed
# for i in range(len(df)):
#     plot_sia_vector(i, df)
````

````python
# prompt: SIA vector pot for single patient circular graph

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Function to plot the SIA vector for a given patient index on a circular plot
def plot_sia_vector_circular(patient_index, df):
    """
    Plots the SIA vector for a single patient on a circular (polar) plot
    using double angle notation.

    Args:
        patient_index (int): The index of the patient in the DataFrame.
        df (pd.DataFrame): The DataFrame containing the SIA data.
    """
    magnitude = df.loc[patient_index, 'SIA-magnitude']
    axis_rad = df.loc[patient_index, 'SIA-axis(radian)']
    axis_deg_adjusted = df.loc[patient_index, 'SIA-axis(degree) adjusted']
    axis_deg_360 = df.loc[patient_index, 'SIA-axis(degree) 0-360']

    # Get the Patient ID if the column exists, otherwise use the index
    if 'Patient ID' in df.columns:
        patient_identifier = df.loc[patient_index, 'Patient ID']
        title_identifier = f"Patient ID: {patient_identifier}"
    else:
        patient_identifier = patient_index
        title_identifier = f"Patient Index: {patient_identifier}"

    # Create the polar plot
    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot the SIA vector as a point and a line from the origin
    # The angle in polar coordinates is the double angle (axis_rad)
    # The radius is the magnitude
    ax.plot([0, axis_rad], [0, magnitude], color='red', linewidth=2, label='SIA Vector')
    ax.plot(axis_rad, magnitude, 'ro') # Plot the tip of the vector

    # Set the direction of 0 degrees (usually East in math, we might want North)
    # ax.set_theta_zero_location("N") # Set 0 degrees to North (up)
    # ax.set_theta_direction(-1) # Set clockwise direction for angles

    # Set the grid lines for angles (every 30 degrees)
    ax.set_xticks(np.linspace(0, 2*np.pi, 12, endpoint=False))
    # Set the labels for the angles (in degrees, matching double angle plot)
    ax.set_xticklabels([f'{int(np.degrees(x)*2) % 360}°' for x in np.linspace(0, 2*np.pi, 12, endpoint=False)]) # Double angle labels
    # You can also set the labels directly based on typical SIA vector analysis
    # ax.set_xticklabels(['0°', '30°', '60°', '90°', '120°', '150°', '180°', '210°', '240°', '270°', '300°', '330°'])


    # Set the radial limits
    # Set r-limits to comfortably fit the vector magnitude
    max_radial = df['SIA-magnitude'].max() * 1.2 # Add some padding
    ax.set_ylim(0, max_radial)

    # Set radial grid lines
    ax.set_yticks(np.linspace(0, max_radial, 5)) # Adjust number of radial ticks as needed
    ax.set_yticklabels([f'{x:.1f}' for x in np.linspace(0, max_radial, 5)]) # Format radial labels

    ax.set_title(f'Circular (Double Angle) SIA Vector Plot for {title_identifier}\nMagnitude vs. Double Angle', va='bottom')

    # Add text label for magnitude and axis
    # Position the text relative to the vector tip or center
    text_angle = axis_rad + np.pi/10 # Offset angle slightly for text
    text_radius = magnitude + max_radial * 0.1 # Offset radius slightly

    # Use annotation for better positioning control
    ax.annotate(f' Mag: {magnitude:.2f}\n Axis (180°): {axis_deg_adjusted:.2f}°\n Axis (360°): {axis_deg_360:.2f}°',
                xy=(axis_rad, magnitude),  # Point to annotate
                xytext=(text_angle, text_radius), # Position of the text
                arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5),
                fontsize=9,
                horizontalalignment='center', verticalalignment='bottom',
                bbox=dict(boxstyle='round,pad=0.5', fc='white', alpha=0.8))


    ax.legend(loc='upper right', bbox_to_anchor=(1.1, 1.1))
    plt.show()

# Example usage: Plot for the first patient (index 0) on a circular plot
plot_sia_vector_circular(0, df)

# Example usage: Plot for the second patient (index 1) on a circular plot
plot_sia_vector_circular(1, df)

# You can loop through a subset of patients if needed
# for i in range(3): # Plotting for the first 3 patients
#     plot_sia_vector_circular(i, df)
````

````python
# prompt: SIA vector plot for 5 patients

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Ensure the 'SIA-axis(degree) 0-360' column exists
# Calculate SIA axis in 0-360 degree range if not already present
if 'SIA-axis(degree) 0-360' not in df.columns:
    df['SIA-axis(degree) 0-360'] = df['SIA-axis(degree)'].apply(lambda x: x + 360 if x < 0 else x)

def plot_sia_vectors_5_patients_circular(df):
    """
    Plots the SIA vectors for the first 5 patients on a single circular plot.

    Args:
        df (pd.DataFrame): The DataFrame containing the SIA data.
    """
    num_patients_to_plot = min(5, len(df)) # Plot up to 5 patients or fewer if less data

    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    colors = plt.cm.get_cmap('viridis', num_patients_to_plot) # Get distinct colors

    for i in range(num_patients_to_plot):
        try:
            magnitude = df.loc[i, 'SIA-magnitude']
            axis_rad = df.loc[i, 'SIA-axis(radian)']
            axis_deg_adjusted = df.loc[i, 'SIA-axis(degree) adjusted']
            axis_deg_360 = df.loc[i, 'SIA-axis(degree) 0-360']

            # Get the Patient ID if the column exists, otherwise use the index
            if 'Patient ID' in df.columns:
                patient_identifier = df.loc[i, 'Patient ID']
            else:
                patient_identifier = i

            # Plot the SIA vector as a line from the origin and a point
            ax.plot([0, axis_rad], [0, magnitude], color=colors(i), linewidth=1.5, label=f'{patient_identifier}')
            ax.plot(axis_rad, magnitude, 'o', color=colors(i), markersize=6) # Plot the tip of the vector

            # Add text label for magnitude and axis near the point
            text_angle = axis_rad # Position text at the vector angle
            text_radius = magnitude + df['SIA-magnitude'].max() * 0.05 # Slightly outside the point
            # Adjust text position slightly based on quadrant for clarity if needed
            ha = 'left' if np.cos(axis_rad) >= 0 else 'right'
            va = 'bottom' if np.sin(axis_rad) >= 0 else 'top'


            ax.text(text_angle, text_radius, f' M:{magnitude:.1f}, A:{axis_deg_adjusted:.0f}°',
                    fontsize=7, color='black', ha=ha, va=va,
                    bbox=dict(boxstyle='round,pad=0.2', fc='white', alpha=0.7))


        except KeyError as e:
            print(f"Error: Required column not found for patient {i} - {e}. Skipping this patient.")
        except IndexError:
            print(f"Error: Patient index {i} is out of bounds. Stopping.")
            break # Stop the loop if index is out of bounds
        except Exception as e:
            print(f"An unexpected error occurred for patient {i}: {e}. Skipping this patient.")


    # Set polar plot properties
    ax.set_theta_zero_location("E") # 0 degrees to the right (East)
    ax.set_theta_direction(1) # Counter-clockwise direction for angles

    # Set the grid lines for angles (every 30 degrees in double angle space)
    ax.set_xticks(np.deg2rad(np.arange(0, 360, 30)/2)) # Set ticks at 0, 15, 30, ..., 165 degrees (double angle)
    # Set the labels for the angles (in degrees, corresponding to double angle)
    ax.set_xticklabels([f'{int(x)}°' for x in np.arange(0, 360, 30)]) # Labels at 0, 30, 60, ..., 330 degrees

    # Set the radial limits based on the maximum magnitude among the plotted patients
    max_radial = df.loc[0:num_patients_to_plot-1, 'SIA-magnitude'].max() * 1.3 # Add some padding
    ax.set_ylim(0, max_radial)

    # Set radial grid lines (adjust number of ticks as needed)
    ax.set_yticks(np.linspace(0, max_radial, 5))
    ax.set_yticklabels([f'{x:.1f}' for x in np.linspace(0, max_radial, 5)]) # Format radial labels

    ax.set_title(f'SIA Vector Plot for {num_patients_to_plot} Patients\nMagnitude vs. Double Angle', va='bottom')

    # Position the legend outside the plot
    ax.legend(loc='upper left', bbox_to_anchor=(1.05, 1))

    plt.show()

# Call the function to plot for the first 5 patients
print("Plotting SIA vectors for the first 5 patients on a single circular plot:")
plot_sia_vectors_5_patients_circular(df)
````

````python
# prompt: Centroid  plot all patients colour

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Assuming df is already loaded and processed as in the preceding code

def plot_sia_vectors_centroid(df):
    """
    Plots the SIA vectors for all patients on a single Cartesian plot
    and also plots the centroid (mean vector).
    Each patient's vector is colored, and the centroid vector is highlighted.

    Args:
        df (pd.DataFrame): The DataFrame containing the SIA data
                           with 'Xsia', 'Ysia', 'SIA-magnitude',
                           'SIA-axis(radian)', and 'SIA-axis(degree) adjusted' columns.
                           'Patient ID' or 'MRD NO' is used for legend if available.
    """
    # Calculate the centroid (mean of Xsia and Ysia)
    centroid_x = df['Xsia'].mean()
    centroid_y = df['Ysia'].mean()

    # Calculate the magnitude and axis of the centroid vector
    centroid_magnitude = np.sqrt(centroid_x**2 + centroid_y**2)
    centroid_axis_rad = np.arctan2(centroid_y, centroid_x)
    centroid_axis_deg_adjusted = np.degrees(centroid_axis_rad)
    centroid_axis_deg_adjusted = centroid_axis_deg_adjusted + 180 if centroid_axis_deg_adjusted < 0 else centroid_axis_deg_adjusted
    centroid_axis_deg_adjusted = centroid_axis_deg_adjusted - 180 if centroid_axis_deg_adjusted >= 180 else centroid_axis_deg_adjusted


    plt.figure(figsize=(8, 8))
    ax = plt.gca()

    # Define a colormap for the patients
    colors = plt.cm.get_cmap('viridis', len(df))

    # Plot each patient's SIA vector
    for i in range(len(df)):
        try:
            sia_x = df.loc[i, 'Xsia']
            sia_y = df.loc[i, 'Ysia']

            # Get patient identifier for legend
            if 'Patient ID' in df.columns:
                patient_identifier = df.loc[i, 'Patient ID']
            elif 'MRD NO' in df.columns:
                 patient_identifier = df.loc[i, 'MRD NO']
            else:
                patient_identifier = i

            # Plot the vector
            ax.quiver(0, 0, sia_x, sia_y, angles='xy', scale_units='xy', scale=1,
                      color=colors(i), alpha=0.6, label=f'{patient_identifier}')

        except KeyError as e:
            print(f"Warning: Skipping patient {i} due to missing column: {e}")
            continue
        except Exception as e:
            print(f"Warning: Skipping patient {i} due to error: {e}")
            continue


    # Plot the centroid vector
    ax.quiver(0, 0, centroid_x, centroid_y, angles='xy', scale_units='xy', scale=1,
              color='black', linewidth=2, label='Centroid Vector (Mean)', zorder=5) # zorder to make it appear on top

    # Add text label for centroid magnitude and axis
    # Position text slightly offset from the centroid vector tip
    text_offset = max(df['SIA-magnitude'].max(), centroid_magnitude) * 0.08 # Adjust offset distance
    centroid_text_x = centroid_x + text_offset * np.cos(centroid_axis_rad)
    centroid_text_y = centroid_y + text_offset * np.sin(centroid_axis_rad)

    # Simple ha/va based on centroid vector quadrant
    ha = 'left' if centroid_x >= 0 else 'right'
    va = 'bottom' if centroid_y >= 0 else 'top'

    ax.text(centroid_text_x, centroid_text_y,
            f' Centroid\n Mag: {centroid_magnitude:.2f} D\n Axis (180°): {centroid_axis_deg_adjusted:.2f}°',
            fontsize=9, color='black', ha=ha, va=va,
            bbox=dict(boxstyle='round,pad=0.3', fc='white', alpha=0.8))


    # Set plot limits based on the maximum SIA magnitude or centroid magnitude
    max_magnitude = max(df['SIA-magnitude'].max(), centroid_magnitude)
    limit = max_magnitude * 1.3 # Add padding for labels
    ax.set_xlim(-limit, limit)
    ax.set_ylim(-limit, limit)

    ax.set_xlabel('SIA X Component (D)')
    ax.set_ylabel('SIA Y Component (D)')
    ax.set_title('SIA Vectors for All Patients and Centroid Vector')
    ax.grid(True)
    ax.axhline(0, color='grey', lw=0.5)
    ax.axvline(0, color='grey', lw=0.5)
    ax.set_aspect('equal', adjustable='box')

    # Create a legend for the patients (optional, can be too crowded)
    # ax.legend(loc='upper right', bbox_to_anchor=(1.2, 1))
    # Or just show the centroid legend
    ax.legend([ax.get_children()[-2]], ['Centroid Vector (Mean)'], loc='upper right')


    plt.show()

# Call the function to plot all patient vectors and the centroid
print("Plotting SIA vectors for all patients and the centroid:")
plot_sia_vectors_centroid(df)
````
