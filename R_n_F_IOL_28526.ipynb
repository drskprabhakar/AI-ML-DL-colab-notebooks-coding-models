### Repeated longitudinal comparisons were performed using repeated-measures methods. Due to >20% missing 3-month observations, primary analyses focused on adequately completed follow-up visits, while 3-month analyses were conducted using complete-case data.

### Shapiro TEST for rigid IOP

```python
from scipy.stats import shapiro

for col in rigid_iop_clean.columns:
    stat, p = shapiro(rigid_iop_clean[col])
    print(col, 'p=', p)
```

### Normality assessment using the Shapiro–Wilk test demonstrated non-normal distribution at POD6wk and POD3 month visits (p<0.05). Therefore, within-group longitudinal comparison of IOP in the rigid IOL group was performed using the Friedman test followed by Wilcoxon signed-rank post-hoc analysis with Bonferroni correction.”


```python
from scipy.stats import friedmanchisquare

stat, p = friedmanchisquare(
    rigid_iop_clean['Preop IOP-R'],
    rigid_iop_clean['POD1 IOP-R '],
    rigid_iop_clean['POD6wk IOP-R '],
    rigid_iop_clean['POD3 mth IOP-R ']
)

print("Friedman statistic =", stat)
print("p value =", p)
```

### Within the rigid IOL group, longitudinal comparison of intraocular pressure across preoperative, postoperative day 1, 6-week, and 3-month visits using the Friedman test showed no statistically significant change over time (χ² = 6.31, p = 0.098).

```python
summary_table = pd.DataFrame(index=rigid_iop_clean.columns)

summary_table['Mean ± SD'] = (
    rigid_iop_clean.mean().round(2).astype(str)
    + ' ± ' +
    rigid_iop_clean.std().round(2).astype(str)
)

summary_table['Median [IQR]'] = (
    rigid_iop_clean.median().round(2).astype(str)
    + ' [' +
    rigid_iop_clean.quantile(0.25).round(2).astype(str)
    + '–' +
    rigid_iop_clean.quantile(0.75).round(2).astype(str)
    + ']'
)

print(summary_table)
```

```python
table = pd.DataFrame({
    'Mean': rigid_iop_clean.mean(),
    'SD': rigid_iop_clean.std(),
    'Median': rigid_iop_clean.median(),
    'Q1': rigid_iop_clean.quantile(0.25),
    'Q3': rigid_iop_clean.quantile(0.75)
})

table['IQR'] = table['Q3'] - table['Q1']

print(table.round(2))
```

### No statistically significant longitudinal change in intraocular pressure was observed within the rigid IOL group across preoperative, POD1, 6-week, and 3-month follow-up visits.”

```python
from scipy.stats import shapiro

# Rigid IOL group
stat_r, p_r = shapiro(rigidiol['Age-R'].dropna())

print("Rigid IOL Age")
print("Shapiro statistic =", round(stat_r,3))
print("p value =", round(p_r,4))

# Foldable IOL group
stat_f, p_f = shapiro(foldableiol['Age (F)'].dropna())

print("\nFoldable IOL Age")
print("Shapiro statistic =", round(stat_f,3))
print("p value =", round(p_f,4))
```

```python
import matplotlib.pyplot as plt
import scipy.stats as stats

fig, axes = plt.subplots(2,2, figsize=(12,10))

# Histogram — Rigid IOL
axes[0,0].hist(rigidiol['Age-R'].dropna(),
               bins=8,
               edgecolor='black')

axes[0,0].set_title('Histogram: Age – Rigid IOL')
axes[0,0].set_xlabel('Age (years)')
axes[0,0].set_ylabel('Frequency')

# Q-Q plot — Rigid IOL
stats.probplot(
    rigidiol['Age-R'].dropna(),
    dist='norm',
    plot=axes[0,1]
)
axes[0,1].set_title('Q–Q Plot: Age – Rigid IOL')

# Histogram — Foldable IOL
axes[1,0].hist(foldableiol['Age (F)'].dropna(),
               bins=8,
               edgecolor='black')

axes[1,0].set_title('Histogram: Age – Foldable IOL')
axes[1,0].set_xlabel('Age (years)')
axes[1,0].set_ylabel('Frequency')

# Q-Q plot — Foldable IOL
stats.probplot(
    foldableiol['Age (F)'].dropna(),
    dist='norm',
    plot=axes[1,1]
)
axes[1,1].set_title('Q–Q Plot: Age – Foldable IOL')

fig.suptitle('Normality Assessment of Age Distribution', fontsize=14)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import ttest_ind

stat, p = ttest_ind(
    rigidiol['Age-R'].dropna(),
    foldableiol['Age (F)'].dropna(),
    equal_var=False      # Welch t-test (recommended)
)

print("t statistic =", round(stat,3))
print("p value =", round(p,4))
```

```python
print("Rigid Age:",
      round(rigidiol['Age-R'].mean(),2),
      "±",
      round(rigidiol['Age-R'].std(),2))

print("Foldable Age:",
      round(foldableiol['Age (F)'].mean(),2),
      "±",
      round(foldableiol['Age (F)'].std(),2))
```

### Age distribution was approximately normal in both groups (Shapiro–Wilk test, p>0.05). The mean age of patients in the rigid IOL group (66.55 ± 9.16 years) was higher than that of the foldable IOL group (62.21 ± 9.03 years); however, the difference was not statistically significant (independent samples t-test, t=1.82, p=0.074).
