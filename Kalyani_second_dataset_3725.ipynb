```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('/content/drive/MyDrive/Dr Kalyani thesis training Dataset KH KV AL ACD LT.xlsx')


```

```python
X = df[['KV', 'KH', 'AGE ']]
y = df['AL']
```

```python
# prompt: Linear regression

import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

# Assuming X and y are already defined and preprocessed as needed

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Create and train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Print the intercept and coefficients
print('Intercept:', model.intercept_)
print('Coefficients:', model.coef_)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print('R-squared:', metrics.r2_score(y_test, y_pred))

# Optional: Visualize the predictions vs actual (for a single feature)
# If you have many features, this visualization is not directly applicable in 2D
# To visualize for a single feature, you would need to select one column from X_test and y_test
# For example, visualizing against 'KV':
# plt.scatter(X_test['KV'], y_test, color='blue')
# plt.plot(X_test['KV'], y_pred, color='red', linewidth=2)
# plt.xlabel('KV')
# plt.ylabel('AL')
# plt.title('Linear Regression (KV vs AL)')
# plt.show()
```

```python
# prompt: OLS

import numpy as np
import statsmodels.api as sm

# Add a constant to the predictor variables for the intercept
X_train_sm = sm.add_constant(X_train)
X_test_sm = sm.add_constant(X_test)

# Create and train the OLS model
ols_model = sm.OLS(y_train, X_train_sm)
ols_results = ols_model.fit()

# Print the OLS summary
print(ols_results.summary())

# Make predictions on the test set
y_pred_sm = ols_results.predict(X_test_sm)

# Evaluate the OLS model
print('\nOLS Model Evaluation:')
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred_sm))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred_sm))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred_sm)))
print('R-squared:', metrics.r2_score(y_test, y_pred_sm))
```

```python
# prompt: Random forest

import numpy as np
from sklearn.ensemble import RandomForestRegressor

# Create and train the Random Forest Regressor model
rf_model = RandomForestRegressor(n_estimators=100, random_state=0) # You can adjust n_estimators
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_rf = rf_model.predict(X_test)

# Evaluate the Random Forest model
print('\nRandom Forest Model Evaluation:')
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred_rf))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred_rf))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred_rf)))
print('R-squared:', metrics.r2_score(y_test, y_pred_rf))

# Optional: Feature importances
print('\nFeature Importances:')
for feature, importance in zip(X.columns, rf_model.feature_importances_):
  print(f'{feature}: {importance}')
```

```python
# prompt: Knn

import numpy as np
from sklearn.neighbors import KNeighborsRegressor

# Create and train the KNN Regressor model
# You can adjust the number of neighbors (n_neighbors)
knn_model = KNeighborsRegressor(n_neighbors=5) # Using 5 neighbors as an example
knn_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_knn = knn_model.predict(X_test)

# Evaluate the KNN model
print('\nKNN Model Evaluation:')
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred_knn))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred_knn))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred_knn)))
print('R-squared:', metrics.r2_score(y_test, y_pred_knn))
```

```python
# prompt: Svm

import numpy as np
from sklearn.svm import SVR

# Create and train the Support Vector Regressor model
# You can experiment with different kernels ('linear', 'poly', 'rbf', 'sigmoid')
# and parameters like C, epsilon, gamma
svm_model = SVR(kernel='rbf', C=1.0, epsilon=0.1)
svm_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_svm = svm_model.predict(X_test)

# Evaluate the SVM model
print('\nSupport Vector Regressor Model Evaluation:')
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred_svm))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred_svm))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred_svm)))
print('R-squared:', metrics.r2_score(y_test, y_pred_svm))
```

```python
# prompt: Decision tree

import numpy as np
from sklearn.tree import DecisionTreeRegressor

# Create and train the Decision Tree Regressor model
# You can experiment with different parameters like max_depth, min_samples_split, etc.
dt_model = DecisionTreeRegressor(random_state=0) # Using default parameters

dt_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_dt = dt_model.predict(X_test)

# Evaluate the Decision Tree model
print('\nDecision Tree Model Evaluation:')
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred_dt))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred_dt))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred_dt)))
print('R-squared:', metrics.r2_score(y_test, y_pred_dt))

# Optional: Feature importances (if applicable to the model)
# Decision trees also provide feature importances
print('\nFeature Importances (Decision Tree):')
for feature, importance in zip(X.columns, dt_model.feature_importances_):
  print(f'{feature}: {importance}')
```

```python
# prompt: Polynomials

import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

# Create polynomial features
# degree=2 means features like x, x^2, xy, y^2, etc.
# include_bias=False removes the constant term, as the linear regression model will add it
poly_features = PolynomialFeatures(degree=2, include_bias=False)

# Create a pipeline that first transforms features to polynomial features and then applies Linear Regression
poly_reg_model = make_pipeline(poly_features, LinearRegression())

# Train the polynomial regression model
poly_reg_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_poly = poly_reg_model.predict(X_test)

# Evaluate the Polynomial Regression model
print('\nPolynomial Regression Model Evaluation (Degree 2):')
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred_poly))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred_poly))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred_poly)))
print('R-squared:', metrics.r2_score(y_test, y_pred_poly))

# Optional: You can access the underlying linear model if needed
# linear_model_in_pipeline = poly_reg_model.named_steps['linearregression']
# print('Polynomial Model Intercept:', linear_model_in_pipeline.intercept_)
# print('Polynomial Model Coefficients:', linear_model_in_pipeline.coef_)
```

```python
# prompt: Pca

import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Assuming X and y are already defined and preprocessed as needed

# Standardize the features before applying PCA
# PCA is affected by scale, so it's good practice to scale your data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA
# Choose the number of components. You can choose a fixed number (e.g., n_components=2)
# or a number of components that explain a certain percentage of the variance (e.g., n_components=0.95)
pca = PCA(n_components=2)  # Let's reduce to 2 principal components for demonstration

# Fit PCA on the scaled training data and transform both train and test sets
# It is important to fit PCA only on the training data to avoid data leakage.
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

pca.fit(X_train_scaled)

X_train_pca = pca.transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

# Print the explained variance ratio for each component
print('Explained variance ratio:', pca.explained_variance_ratio_)
print('Total explained variance:', sum(pca.explained_variance_ratio_))

# Print the principal components (eigenvectors)
print('Principal Components (Eigenvectors):\n', pca.components_)

# Now you can use the PCA-transformed data (X_train_pca, X_test_pca) to train your regression models
# For example, training a Linear Regression model on the PCA-transformed data:
print('\nLinear Regression on PCA-transformed data:')
model_pca = LinearRegression()
model_pca.fit(X_train_pca, y_train)

# Make predictions on the PCA-transformed test set
y_pred_pca = model_pca.predict(X_test_pca)

# Evaluate the model
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred_pca))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred_pca))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred_pca)))
print('R-squared:', metrics.r2_score(y_test, y_pred_pca))

# You can apply PCA to other models as well by replacing X_train and X_test with X_train_pca and X_test_pca
# For example, training a Random Forest model on PCA-transformed data:
# print('\nRandom Forest on PCA-transformed data:')
# rf_model_pca = RandomForestRegressor(n_estimators=100, random_state=0)
# rf_model_pca.fit(X_train_pca, y_train)
# y_pred_rf_pca = rf_model_pca.predict(X_test_pca)
# print('R-squared:', metrics.r2_score(y_test, y_pred_rf_pca))
```
