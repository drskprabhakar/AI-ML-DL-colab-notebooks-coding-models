```python
df = pd.read_excel('/content/drive/MyDrive/Dr Aishwarya thesis excel sheet cleaned.xlsx')
```

```python
# find preop magnitude(m1) and postop magnitude(m2)
df['m1'] = df['pre-op KV']-df['pre-op KH']
df['m2'] = df['post-op KV']-df['post-op KH']
df
```

```python
# define a1,a2
df['a1'] = df['pre-op Axis']
df['a2'] = df['post-op Axis']
df
```

```python
# find x1,y1,x2,y2

df['x1'] = df['m1'] * np.cos(df['a1'] * 2)
df['y1'] = df['m1'] * np.sin(df['a1'] * 2)
df['x2'] = df['m2'] * np.cos(df['a2'] * 2)
df['y2'] = df['m2'] * np.sin(df['a2'] * 2)
df
```

```python
# find Xsia and Ysia
df['Xsia'] = df['x2'] - df['x1']
df['Ysia'] = df['y2'] - df['y1']
df
```

```python
# find SIA vector magnitude

df['SIA-vector-magnitude'] = np.sqrt(df['Xsia'] ** 2 + df['Ysia'] ** 2)
df
```

```python
# find SIA-vector-axis(radian)

df['SIA-vector-axis'] = np.arctan2(df['Ysia'], df['Xsia'])
df['SIA-vector-axis'] = df['SIA-vector-axis'] * 180 /np.pi
df
```

```python
# prompt: plot SIA-vector for single patient

import pandas as pd
import matplotlib.pyplot as plt
def plot_sia_vector(df, patient_index):
    """
    Plots the SIA vector for a single patient.

    Args:
        df (pd.DataFrame): The DataFrame containing SIA data.
        patient_index (int): The index of the patient in the DataFrame.
    """
    patient_data = df.iloc[patient_index]

    x_sia = patient_data['Xsia']
    y_sia = patient_data['Ysia']
    sia_magnitude = patient_data['SIA-vector-magnitude']
    sia_axis_degrees = patient_data['SIA-vector-axis']

    plt.figure(figsize=(6, 6))
    plt.quiver(0, 0, x_sia, y_sia, angles='xy', scale_units='xy', scale=1, color='b', label='SIA Vector')

    plt.xlim(-1 * max(abs(x_sia), abs(y_sia)) - 0.5, max(abs(x_sia), abs(y_sia)) + 0.5)
    plt.ylim(-1 * max(abs(x_sia), abs(y_sia)) - 0.5, max(abs(x_sia), abs(y_sia)) + 0.5)

    plt.axhline(0, color='grey', lw=0.5)
    plt.axvline(0, color='grey', lw=0.5)

    plt.title(f'SIA Vector for Patient Index {patient_index}')
    plt.xlabel('X_SIA')
    plt.ylabel('Y_SIA')
    plt.grid(True)
    plt.gca().set_aspect('equal', adjustable='box')

    # Add text annotation for magnitude and axis
    plt.text(x_sia, y_sia, f' Magnitude: {sia_magnitude:.2f}\n Axis: {sia_axis_degrees:.2f}°',
             fontsize=9, ha='left', va='bottom')

    plt.legend()
    plt.show()

# Example usage: Plot the SIA vector for the first patient (index 0)
plot_sia_vector(df, 0)

# You can change the index to plot for a different patient
# plot_sia_vector(df, 5)
```

```python
# prompt: vector plots describe

import matplotlib.pyplot as plt
The preceding code calculates and visualizes the Surgically Induced Astigmatism (SIA) vector for patients based on their pre-operative and post-operative keratometry (KV, KH, and Axis) values.

Here's a breakdown of what the vector plots describe:

1.  **Understanding SIA as a Vector:** Astigmatism can be represented as a vector in a two-dimensional space (often called double-angle vector space or polar space). The pre-operative and post-operative astigmatism states are treated as vectors. The difference between the post-operative vector and the pre-operative vector gives the SIA vector. This vector has both a magnitude (how much the astigmatism changed) and an axis (the direction of that change).

2.  **Calculation Steps:**
    *   `m1`, `m2`, `a1`, `a2`: These represent the magnitude and axis of pre-op and post-op astigmatism. The magnitude is the difference between the principal meridians (KV - KH). The axis is the measured axis.
    *   `x1`, `y1`, `x2`, `y2`: The magnitudes and axes are converted into their Cartesian components (`x` and `y`) in the double-angle vector space. The formula `m * cos(2*axis)` and `m * sin(2*axis)` is used for this conversion. The axis is multiplied by 2 because astigmatism vectors repeat every 180 degrees.
    *   `Xsia`, `Ysia`: These are the Cartesian components of the SIA vector, calculated by subtracting the pre-op components from the post-op components (`x2 - x1`, `y2 - y1`).
    *   `SIA-vector-magnitude`: This is the length of the SIA vector, calculated using the Pythagorean theorem (`sqrt(Xsia^2 + Ysia^2)`). It represents the amount of astigmatic change induced by the surgery.
    *   `SIA-vector-axis`: This is the angle of the SIA vector in the double-angle space, calculated using `arctan2(Ysia, Xsia)`. `arctan2` is used to handle all four quadrants correctly. The resulting angle is then converted from radians to degrees and adjusted to be between 0 and 180 degrees for easier interpretation in clinical context (where astigmatism axis is typically represented between 0 and 180).

3.  **Vector Plots Description:**

    *   **`plot_sia_vector(df, patient_index)` (Cartesian Plot):** This function plots a single SIA vector for a specific patient in a standard Cartesian coordinate system (X_SIA vs Y_SIA).
        *   It uses `plt.quiver(0, 0, x_sia, y_sia, ...)` to draw an arrow starting from the origin (0,0) and ending at the point (X_SIA, Y_SIA).
        *   This plot directly visualizes the `Xsia` and `Ysia` components of the SIA.
        *   It includes annotations for the calculated `SIA-vector-magnitude` and `SIA-vector-axis` for that specific patient, making it easy to see the resultant vector and its properties.
        *   The axes are scaled to maintain an equal aspect ratio, which is important for correctly representing vectors.

    *   **Polar Plot of SIA Vectors (for multiple patients):** This plot visualizes multiple SIA vectors (up to 5 in the provided code) in a polar coordinate system.
        *   The angle (`theta`) in the polar plot represents the `SIA-vector-axis` (converted to radians and adjusted).
        *   The radial distance (`r`) from the center represents the `SIA-vector-magnitude`.
        *   Each patient's SIA vector is shown as a line segment originating from the center (the "pole") and extending outwards to the point defined by their magnitude and axis.
        *   This plot is particularly useful for visualizing the *distribution* of SIA vectors across a group of patients. It quickly shows if the induced astigmatism is predominantly in a certain direction or if the magnitudes are clustered around certain values. Setting the theta zero location to "N" (North/Top) and direction clockwise is conventional for ophthalmological astigmatism plots.

    *   **`plot_single_patient_sia_analysis(df, patient_index)` (Polar Plot with Details):** This function provides a more detailed analysis for a single patient, including printing the calculation steps and then plotting the final SIA vector in a polar coordinate system.
        *   Similar to the multi-patient polar plot, it shows the SIA vector using the `SIA-vector-axis` as the angle and `SIA-vector-magnitude` as the radial distance.
        *   It uses `ax.plot([0, sia_axis_radians], [0, sia_magnitude], ...)` to draw the vector from the origin.
        *   It adds an annotation near the end of the vector indicating the magnitude and axis, similar to the Cartesian single-patient plot.
        *   It dynamically sets the limits of the radial axis based on the SIA magnitude for better scaling.

In summary, the vector plots describe the **direction and magnitude of the astigmatism change induced by the surgical procedure** for individual patients or visualize the pattern of these changes across a group of patients, using the SIA vector components calculated in the code.
```

```python
# prompt:  centroid vector plot color map

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
def plot_centroid_vector_with_colormap(df):
    """
    Calculates and plots the centroid SIA vector for the entire dataset,
    with individual patient vectors colored by their magnitude.

    Args:
        df (pd.DataFrame): The DataFrame containing SIA data with 'Xsia' and 'Ysia' columns.
    """
    # Calculate the centroid vector
    centroid_Xsia = df['Xsia'].mean()
    centroid_Ysia = df['Ysia'].mean()
    centroid_magnitude = np.sqrt(centroid_Xsia**2 + centroid_Ysia**2)
    centroid_axis_radians = np.arctan2(centroid_Ysia, centroid_Xsia)
    centroid_axis_degrees = np.rad2deg(centroid_axis_radians)
    centroid_axis_degrees = centroid_axis_degrees + 180 if centroid_axis_degrees < 0 else centroid_axis_degrees # Quadrantic handling

    print(f"--- Centroid SIA Vector ---")
    print(f"X_SIA (Mean): {centroid_Xsia:.2f}")
    print(f"Y_SIA (Mean): {centroid_Ysia:.2f}")
    print(f"Centroid Magnitude: {centroid_magnitude:.2f}")
    print(f"Centroid Axis: {centroid_axis_degrees:.2f}°")
    print("--------------------------")

    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot individual patient vectors colored by magnitude
    # Use a colormap
    cmap = plt.cm.viridis # You can choose other colormaps like 'plasma', 'inferno', 'cividis', 'magma'
    norm = plt.Normalize(df['SIA-vector-magnitude'].min(), df['SIA-vector-magnitude'].max())

    for i in range(len(df)):
        sia_magnitude = df.iloc[i]['SIA-vector-magnitude']
        sia_axis_degrees = df.iloc[i]['SIA-vector-axis']
        sia_axis_radians = np.deg2rad(sia_axis_degrees)

        # Get color based on magnitude
        color = cmap(norm(sia_magnitude))

        ax.plot([0, sia_axis_radians], [0, sia_magnitude], color=color, linestyle='-', linewidth=0.5, alpha=0.6) # Plot as lines

    # Plot the centroid vector (thicker line)
    ax.plot([0, centroid_axis_radians], [0, centroid_magnitude], color='red', marker='o', linestyle='-', linewidth=3, label='Centroid Vector')


    ax.set_theta_zero_location("N") # Set 0 degrees to the top (North)
    ax.set_theta_direction(-1)     # Set angle direction clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title("SIA Vectors and Centroid Vector (Colored by Magnitude)", va='bottom')

    # Set r-limits dynamically
    max_r = max(df['SIA-vector-magnitude'].max(), centroid_magnitude) + 1 # Add some padding
    ax.set_rlim(0, max_r)
    ax.set_rticks(np.linspace(0, max_r, 5)) # Adjust radial ticks

    # Add color bar for magnitude
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([]) # You can set an array here, but an empty one works for just showing the color bar
    cbar = plt.colorbar(sm, ax=ax, label='SIA Vector Magnitude')


    # Add grid
    ax.grid(True)

    ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5)) # Place legend outside the plot


    plt.show()

# Call the function to plot the centroid vector with colormap
plot_centroid_vector_with_colormap(df)
```

```python
# prompt: any other

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
def plot_target_induced_astigmatism_vectors(df, target_magnitude, target_axis_degrees):
    """
    Calculates and plots the SIA vectors relative to a target induced astigmatism vector.

    Args:
        df (pd.DataFrame): The DataFrame containing SIA data.
        target_magnitude (float): The target SIA magnitude.
        target_axis_degrees (float): The target SIA axis in degrees (0-180).
    """
    # Convert target vector to Cartesian components
    target_axis_radians = np.deg2rad(target_axis_degrees)
    target_X = target_magnitude * np.cos(target_axis_radians * 2)
    target_Y = target_magnitude * np.sin(target_axis_radians * 2)

    print(f"--- Target Induced Astigmatism ---")
    print(f"Target Magnitude: {target_magnitude:.2f}, Target Axis: {target_axis_degrees:.2f}°")
    print(f"Target X (Double Angle): {target_X:.2f}, Target Y (Double Angle): {target_Y:.2f}")
    print("----------------------------------")

    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot the target vector
    ax.plot([0, target_axis_radians], [0, target_magnitude], color='black', marker='*', markersize=10, linestyle='--', linewidth=2, label='Target SIA Vector')

    # Plot individual patient SIA vectors
    cmap = plt.cm.viridis # Colormap for individual vectors
    norm = plt.Normalize(df['SIA-vector-magnitude'].min(), df['SIA-vector-magnitude'].max())

    for i in range(len(df)):
        sia_magnitude = df.iloc[i]['SIA-vector-magnitude']
        sia_axis_degrees = df.iloc[i]['SIA-vector-axis']
        sia_axis_radians = np.deg2rad(sia_axis_degrees)

        color = cmap(norm(sia_magnitude))

        ax.plot([0, sia_axis_radians], [0, sia_magnitude], color=color, linestyle='-', linewidth=0.5, alpha=0.6) # Plot as lines

    ax.set_theta_zero_location("N") # Set 0 degrees to the top (North)
    ax.set_theta_direction(-1)     # Set angle direction clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title("SIA Vectors Relative to Target Vector (Colored by Magnitude)", va='bottom')

    # Set r-limits dynamically
    max_r = max(df['SIA-vector-magnitude'].max(), target_magnitude) + 1 # Add some padding
    ax.set_rlim(0, max_r)
    ax.set_rticks(np.linspace(0, max_r, 5)) # Adjust radial ticks

    # Add color bar for magnitude
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = plt.colorbar(sm, ax=ax, label='SIA Vector Magnitude')

    # Add grid
    ax.grid(True)

    ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5))

    plt.show()

# Example usage: Assuming a target SIA magnitude of 0.5 D at an axis of 90 degrees
target_magnitude_example = 0.5
target_axis_degrees_example = 90

plot_target_induced_astigmatism_vectors(df, target_magnitude_example, target_axis_degrees_example)
```

```python
# prompt: plot double angle vector

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
def plot_double_angle_vectors_with_labels(df, num_patients=5):
    """
    Plots pre-operative, post-operative, and SIA vectors for a specified number of patients
    in the double-angle vector space (polar plot) with labels.

    Args:
        df (pd.DataFrame): The DataFrame containing pre-op, post-op, and SIA vector data.
        num_patients (int): The number of patients to plot (default is 5).
    """
    plt.figure(figsize=(10, 10))
    ax = plt.subplot(111, projection='polar')

    # Get unique patient indices to plot
    patient_indices_to_plot = df.index[:min(num_patients, len(df))]

    # Define distinct colors for each patient
    # Using a larger colormap like 'tab20' or 'viridis' scaled to the number of patients
    cmap_patients = plt.cm.get_cmap('tab20', len(patient_indices_to_plot))

    # Define colors for vector types (pre-op, post-op, SIA)
    color_preop = 'blue'
    color_postop = 'green'
    color_sia = 'red'

    max_magnitude_overall = 0

    for i, patient_index in enumerate(patient_indices_to_plot):
        patient_data = df.loc[patient_index]

        # Pre-op Vector
        m1 = patient_data['m1']
        a1_degrees = patient_data['a1']
        a1_radians_double = np.deg2rad(a1_degrees * 2) # Double angle for plotting

        # Post-op Vector
        m2 = patient_data['m2']
        a2_degrees = patient_data['a2']
        a2_radians_double = np.deg2rad(a2_degrees * 2) # Double angle for plotting

        # SIA Vector
        sia_magnitude = patient_data['SIA-vector-magnitude']
        sia_axis_degrees = patient_data['SIA-vector-axis'] # Already 0-180
        sia_axis_radians_double = np.deg2rad(sia_axis_degrees * 2) # Double angle for plotting


        # Plot pre-op vector
        ax.plot([0, a1_radians_double], [0, m1], color=color_preop, linestyle='-', linewidth=1, alpha=0.7, label=f'Patient {patient_index} Pre-op' if i==0 else "")

        # Plot post-op vector
        ax.plot([0, a2_radians_double], [0, m2], color=color_postop, linestyle='-', linewidth=1, alpha=0.7, label=f'Patient {patient_index} Post-op' if i==0 else "")

        # Plot SIA vector (from the end of pre-op to the end of post-op) - This is the vector addition
        # Plotting the SIA vector from the origin as a separate arrow is also common.
        # Let's plot it from the origin in double-angle space for clarity
        ax.plot([0, sia_axis_radians_double], [0, sia_magnitude], color=color_sia, linestyle='--', linewidth=1.5, alpha=0.8, label=f'Patient {patient_index} SIA' if i==0 else "")


        # Update max magnitude for setting limits
        max_magnitude_overall = max(max_magnitude_overall, m1, m2, sia_magnitude)

    # Set properties for the polar plot
    ax.set_theta_zero_location("E") # Set 0 degrees to the right (often used in double angle plots)
    ax.set_theta_direction(1)     # Set angle direction counter-clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title(f"Double Angle Vectors for {min(num_patients, len(df))} Patients", va='bottom')

    # Set r-limits dynamically
    plot_limit = max_magnitude_overall + 0.5 # Add some padding
    ax.set_rlim(0, plot_limit)
    ax.set_rticks(np.linspace(0, plot_limit, 5)) # Adjust radial ticks

    # Add grid
    ax.grid(True)

    # Create custom legend handles for vector types
    preop_legend = plt.Line2D([0], [0], color=color_preop, lw=1, label='Pre-op Vector')
    postop_legend = plt.Line2D([0], [0], color=color_postop, lw=1, label='Post-op Vector')
    sia_legend = plt.Line2D([0], [0], color=color_sia, lw=1.5, linestyle='--', label='SIA Vector')

    # Combine the custom legend handles
    legend_handles = [preop_legend, postop_legend, sia_legend]

    # Add legend
    ax.legend(handles=legend_handles, loc='lower left', bbox_to_anchor=(1.05, 0.5)) # Place legend outside the plot


    plt.show()

# Example usage: Plot the double angle vectors for the first 5 patients
plot_double_angle_vectors_with_labels(df, num_patients=5)

# You can change the number of patients
# plot_double_angle_vectors_with_labels(df, num_patients=10)
```

```python
# prompt: any other

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
def plot_vector_addition_in_double_angle_space(df, patient_index):
    """
    Illustrates the vector addition (pre-op + SIA = post-op) in the double-angle vector space
    for a single patient.

    Args:
        df (pd.DataFrame): The DataFrame containing pre-op, post-op, and SIA vector data.
        patient_index (int): The index of the patient in the DataFrame.
    """
    if patient_index < 0 or patient_index >= len(df):
        print(f"Error: Patient index {patient_index} is out of bounds.")
        return

    patient_data = df.iloc[patient_index]

    # Get vector components
    m1 = patient_data['m1']
    a1_degrees = patient_data['a1']
    a1_radians_double = np.deg2rad(a1_degrees * 2)

    m2 = patient_data['m2']
    a2_degrees = patient_data['a2']
    a2_radians_double = np.deg2rad(a2_degrees * 2)

    sia_magnitude = patient_data['SIA-vector-magnitude']
    sia_axis_degrees = patient_data['SIA-vector-axis']
    sia_axis_radians_double = np.deg2rad(sia_axis_degrees * 2)

    # Convert pre-op, post-op, and SIA to Cartesian coordinates in double-angle space
    x1 = patient_data['x1']
    y1 = patient_data['y1']

    x2 = patient_data['x2']
    y2 = patient_data['y2']

    xsia = patient_data['Xsia']
    ysia = patient_data['Ysia']


    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot Pre-op vector from origin
    ax.plot([0, a1_radians_double], [0, m1], color='blue', marker='o', markersize=5, linestyle='-', linewidth=2, label='Pre-op Vector')
    ax.annotate(f'Pre-op\nM={m1:.2f}\nA={a1_degrees:.0f}°',
                xy=(a1_radians_double, m1), xycoords='data',
                xytext=(5, 5), textcoords='offset points',
                fontsize=9, ha='left', va='bottom', color='blue')

    # Plot SIA vector starting from the end of the Pre-op vector (vector addition)
    # The end point of the pre-op vector is (x1, y1) in Cartesian
    # The SIA vector in Cartesian is (xsia, ysia)
    # The end point of the SIA vector added to pre-op is (x1 + xsia, y1 + ysia) which should equal (x2, y2)
    # In polar coordinates, we draw the SIA vector starting from the (x1, y1) location.
    # We need to convert (x1, y1) to polar coordinates to start drawing.
    # The SIA vector is (xsia, ysia) in Cartesian. Its magnitude is sia_magnitude and double angle is sia_axis_radians_double.
    # To plot from (x1, y1), we translate the origin.
    # The end point will be at (x1 + xsia, y1 + ysia).
    # We can draw an arrow from (x1, y1) to (x1+xsia, y1+ysia) in Cartesian and then convert to polar.
    # A simpler way in polar plot is to draw the SIA vector from (x1,y1) by defining the start and end points in Cartesian and using annotate or drawing lines.

    # Let's draw the SIA vector as a dotted line from the end of pre-op to the end of post-op
    # Start point of SIA vector in Cartesian is (x1, y1)
    # End point of SIA vector in Cartesian is (x2, y2)
    ax.plot([np.arctan2(y1, x1), np.arctan2(y2, x2)], [np.sqrt(x1**2 + y1**2), np.sqrt(x2**2 + y2**2)],
            color='red', linestyle=':', linewidth=2, alpha=0.8)

    # Alternatively, plot SIA from the origin in double-angle space
    ax.plot([0, sia_axis_radians_double], [0, sia_magnitude], color='red', marker='x', markersize=5, linestyle='--', linewidth=2, label='SIA Vector')
    ax.annotate(f'SIA\nM={sia_magnitude:.2f}\nA={sia_axis_degrees:.0f}°',
                xy=(sia_axis_radians_double, sia_magnitude), xycoords='data',
                xytext=(-5, -5), textcoords='offset points', ha='right', va='top',
                fontsize=9, color='red')


    # Plot Post-op vector from origin
    ax.plot([0, a2_radians_double], [0, m2], color='green', marker='^', markersize=5, linestyle='-', linewidth=2, label='Post-op Vector')
    ax.annotate(f'Post-op\nM={m2:.2f}\nA={a2_degrees:.0f}°',
                xy=(a2_radians_double, m2), xycoords='data',
                xytext=(5, 5), textcoords='offset points',
                fontsize=9, ha='left', va='bottom', color='green')


    # Set properties for the polar plot
    ax.set_theta_zero_location("E") # Set 0 degrees to the right (often used in double angle plots)
    ax.set_theta_direction(1)     # Set angle direction counter-clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title(f"Vector Addition: Pre-op + SIA = Post-op (Patient Index {patient_index})", va='bottom')

    # Set r-limits dynamically
    max_magnitude_overall = max(m1, m2, sia_magnitude)
    plot_limit = max_magnitude_overall + 0.5 # Add some padding
    ax.set_rlim(0, plot_limit)
    ax.set_rticks(np.linspace(0, plot_limit, 5)) # Adjust radial ticks

    # Add grid
    ax.grid(True)

    # Add legend
    ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5))

    plt.show()

# Example usage: Plot the vector addition for the first patient (index 0)
plot_vector_addition_in_double_angle_space(df, 0)

# You can change the index to visualize for a different patient
# plot_vector_addition_in_double_angle_space(df, 5)


def plot_surgically_induced_astigmatism_analysis(df, patient_index=None, num_patients_polar=5, target_magnitude=None, target_axis_degrees=None):
    """
    Provides a comprehensive analysis and visualization of SIA.

    Args:
        df (pd.DataFrame): The DataFrame containing calculated SIA data.
        patient_index (int, optional): Index of a single patient to focus on for detailed plots.
                                       If None, only summary plots are generated.
        num_patients_polar (int): Number of individual patient vectors to show in the polar plots
                                  if patient_index is None. Defaults to 5.
        target_magnitude (float, optional): Target SIA magnitude for target vector plot.
        target_axis_degrees (float, optional): Target SIA axis in degrees for target vector plot.
    """

    print("--- Surgically Induced Astigmatism (SIA) Analysis ---")

    # 1. Summary Statistics and Histograms
    print("\nSummary Statistics:")
    print(df[['SIA-vector-magnitude', 'SIA-vector-axis']].describe())

    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    df['SIA-vector-magnitude'].hist(bins=20)
    plt.title('Histogram of SIA Vector Magnitude')
    plt.xlabel('SIA Vector Magnitude (Diopters)')
    plt.ylabel('Frequency')

    plt.subplot(1, 2, 2)
    df['SIA-vector-axis'].hist(bins=40)
    plt.title('Histogram of SIA Vector Axis')
    plt.xlabel('SIA Vector Axis (Degrees)')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()

    # 2. Scatter plot of Magnitude vs Axis
    plt.figure(figsize=(8, 6))
    plt.scatter(df['SIA-vector-axis'], df['SIA-vector-magnitude'], alpha=0.6)
    plt.title('SIA Vector Magnitude vs. SIA Vector Axis')
    plt.xlabel('SIA Vector Axis (Degrees)')
    plt.ylabel('SIA Vector Magnitude (Diopters)')
    plt.grid(True)
    plt.show()

    # 3. Overall SIA Vector Distribution (Polar Plot)
    print("\nSIA Vector Distribution (Polar Plot):")
    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    cmap = plt.cm.viridis
    norm = plt.Normalize(df['SIA-vector-magnitude'].min(), df['SIA-vector-magnitude'].max())

    for i in range(len(df)):
        sia_magnitude = df.iloc[i]['SIA-vector-magnitude']
        sia_axis_degrees = df.iloc[i]['SIA-vector-axis']
        sia_axis_radians = np.deg2rad(sia_axis_degrees)

        color = cmap(norm(sia_magnitude))

        ax.plot([0, sia_axis_radians], [0, sia_magnitude], color=color, linestyle='-', linewidth=0.7, alpha=0.6)

    ax.set_theta_zero_location("N") # 0 degrees at top
    ax.set_theta_direction(-1)     # Clockwise
    ax.set_rlabel_position(-22.5)
    ax.set_title("Distribution of Individual SIA Vectors (Colored by Magnitude)", va='bottom')

    max_r = df['SIA-vector-magnitude'].max() + 0.5
    ax.set_rlim(0, max_r)
    ax.set_rticks(np.linspace(0, max_r, 5))

    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = plt.colorbar(sm, ax=ax, label='SIA Vector Magnitude')

    ax.grid(True)
    plt.show()

    # 4. Centroid SIA Vector
    centroid_Xsia = df['Xsia'].mean()
    centroid_Ysia = df['Ysia'].mean()
    centroid_magnitude = np.sqrt(centroid_Xsia**2 + centroid_Ysia**2)
    centroid_axis_radians = np.arctan2(centroid_Ysia, centroid_Xsia)
    centroid_axis_degrees = np.rad2deg(centroid_axis_radians)
    centroid_axis_degrees = centroid_axis_degrees + 180 if centroid_axis_degrees < 0 else centroid_axis_degrees

    print("\nCentroid SIA Vector for the Group:")
    print(f"  Magnitude: {centroid_magnitude:.2f} Diopters")
    print(f"  Axis: {centroid_axis_degrees:.2f} Degrees")

    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot the centroid vector
    ax.plot([0, np.deg2rad(centroid_axis_degrees)], [0, centroid_magnitude], color='red', marker='o', markersize=8, linestyle='-', linewidth=3, label='Centroid SIA Vector')

    ax.set_theta_zero_location("N") # 0 degrees at top
    ax.set_theta_direction(-1)     # Clockwise
    ax.set_rlabel_position(-22.5)
    ax.set_title("Centroid SIA Vector", va='bottom')

    max_r = max(df['SIA-vector-magnitude'].max(), centroid_magnitude) + 0.5
    ax.set_rlim(0, max_r)
    ax.set_rticks(np.linspace(0, max_r, 5))

    ax.grid(True)
    ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5))
    plt.show()


    # 5. Target Induced Astigmatism (if specified)
    if target_magnitude is not None and target_axis_degrees is not None:
         print("\nSIA Vectors Relative to Target Induced Astigmatism:")
         plot_target_induced_astigmatism_vectors(df, target_magnitude, target_axis_degrees)


    # 6. Individual Patient Analysis (if index is provided)
    if patient_index is not None:
        print(f"\n--- Detailed Analysis for Patient Index: {patient_index} ---")
        if patient_index < 0 or patient_index >= len(df):
            print(f"Error: Patient index {patient_index} is out of bounds.")
        else:
            patient_data = df.iloc[patient_index]
            print(f"Pre-op (M, A): ({patient_data['m1']:.2f}, {patient_data['a1']:.0f}°)")
            print(f"Post-op (M, A): ({patient_data['m2']:.2f}, {patient_data['a2']:.0f}°)")
            print(f"SIA Vector (Magnitude, Axis): ({patient_data['SIA-vector-magnitude']:.2f}, {patient_data['SIA-vector-axis']:.0f}°)")
            print(f"SIA Vector (X, Y): ({patient_data['Xsia']:.2f}, {patient_data['Ysia']:.2f})")

            # Plot individual SIA vector (Cartesian)
            plot_sia_vector_with_legend(df, patient_index)

            # Plot vector addition in double-angle space (Polar)
            plot_vector_addition_in_double_angle_space(df, patient_index)

    else: # If no specific patient index, plot individual vectors for a few patients in polar
         print(f"\nIndividual SIA Vectors for First {num_patients_polar} Patients (Polar Plot):")
         plot_polar_sia_vectors_with_legend(df, num_patients=num_patients_polar)


    print("\n--- Analysis Complete ---")


# Example Usage of the Comprehensive Function:

# To run the full analysis including summary plots and centroid:
plot_surgically_induced_astigmatism_analysis(df)

# To run the full analysis focusing on a specific patient (e.g., index 5):
# plot_surgically_induced_astigmatism_analysis(df, patient_index=5)

# To run the full analysis and plot relative to a target vector (e.g., 0.5 D @ 90 degrees):
# plot_surgically_induced_astigmatism_analysis(df, target_magnitude=0.5, target_axis_degrees=90)

# To run the full analysis focusing on a patient AND relative to a target vector:
# plot_surgically_induced_astigmatism_analysis(df, patient_index=5, target_magnitude=0.5, target_axis_degrees=90)
```

```python
# prompt: SIA axis plot

Here is a description of what SIA axis plot describes:

**Surgically Induced Astigmatism (SIA) Axis Plot Description**

SIA axis plots are visualizations used in ophthalmology, particularly in corneal and cataract surgery, to analyze the magnitude and direction of astigmatic change induced by a surgical procedure. Astigmatism is often treated as a vector quantity, and SIA represents the vector difference between the post-operative astigmatism and the pre-operative astigmatism.

The plots generated in the provided code describe the following aspects of SIA:

1.  **Individual SIA Vector Distribution (Polar Plot):**
    *   This plot shows each patient's SIA vector as an arrow originating from the center (the origin or "pole") of a polar coordinate system.
    *   The **angle** of the vector (measured from a reference direction, often 0 degrees vertically upwards or horizontally to the right, depending on convention) represents the **axis of the surgically induced astigmatism**. In the ophthalmological context, axes are typically represented between 0 and 180 degrees. The double-angle transformation is used in the calculations because astigmatism vectors repeat every 180 degrees.
    *   The **length** (or magnitude) of the vector represents the **amount of astigmatism induced by the surgery**, typically measured in Diopters (D).
    *   Plotting all individual SIA vectors together provides a visual representation of the spread and clustering of surgical outcomes across the patient group. Vectors pointing in similar directions indicate consistent SIA axes, while the lengths indicate the variability in the induced magnitude. Coloring the vectors by magnitude helps to see if larger SIA values are associated with specific axes.

2.  **Centroid SIA Vector Plot (Polar Plot):**
    *   This plot shows a single vector representing the **average SIA** for the entire group of patients.
    *   It is calculated by averaging the X and Y components of all individual SIA vectors in the double-angle vector space.
    *   The centroid vector's magnitude and axis represent the overall tendency of the surgery to induce a certain amount and direction of astigmatism.
    *   This is a key metric for evaluating the *predictability* and *effectiveness* of a surgical technique or a cohort's overall SIA response. A small centroid magnitude suggests the surgery is highly predictable and induces little overall astigmatism on average, while a larger centroid magnitude and a consistent axis indicate a predictable induction of astigmatism in that direction.

3.  **Single Patient SIA Vector Plot (Cartesian Plot):**
    *   This plot focuses on a single patient and shows their SIA vector in a standard Cartesian coordinate system.
    *   The x-axis represents the X_SIA component and the y-axis represents the Y_SIA component of the SIA vector in double-angle space.
    *   The arrow starts from the origin (0,0) and ends at the point (X_SIA, Y_SIA).
    *   This plot directly visualizes the components of the SIA vector and often includes annotations for the resulting magnitude and axis in clinical terms (Diopters and 0-180 degrees). It's useful for understanding the SIA of an individual case in terms of its horizontal and vertical components in the double-angle space.

4.  **Vector Addition Plot (Pre-op + SIA = Post-op) (Polar Plot):**
    *   This plot demonstrates the fundamental vector principle: the pre-operative astigmatism vector plus the SIA vector equals the post-operative astigmatism vector.
    *   It typically shows the pre-op vector from the origin, followed by the SIA vector starting from the *end* of the pre-op vector, and finally the post-op vector originating from the origin and ending at the *end* of the SIA vector (which should coincide with the end of the vector chain).
    *   This visualization helps confirm that the SIA calculation correctly captures the change from the pre-operative state to the post-operative state as a vector addition in the double-angle space. It illustrates how the induced astigmatism interacts with the existing astigmatism.

In essence, these plots translate the changes in corneal shape (represented by astigmatism) caused by surgery into a clear visual format using vectors, making it easier to understand the magnitude, direction, and variability of the surgically induced astigmatism for individual patients and the patient group as a whole. They are crucial for surgical planning, outcome analysis, and comparing different surgical techniques or technologies.
```

```python
# prompt: any other

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Calculate Magnitude of Error (ME)
# ME is the magnitude of the SIA vector
df['Magnitude of Error (ME)'] = df['SIA-vector-magnitude']

# Calculate Angle of Error (AE)
# AE is the angle of the SIA vector relative to the 0/180 axis
# Since SIA-vector-axis is already calculated and adjusted to be between 0 and 180,
# the AE is simply the SIA-vector-axis itself.
# Note: This is a simplified definition of AE. More complex definitions might consider
# the difference from a target induced astigmatism axis or the axis of the pre-op astigmatism.
# Based on the context of calculating SIA as the difference vector, using the SIA vector axis
# as the Angle of Error is a common approach.
df['Angle of Error (AE)'] = df['SIA-vector-axis']

print("Magnitude of Error (ME) and Angle of Error (AE) calculated.")
print(df[['Magnitude of Error (ME)', 'Angle of Error (AE)']].head())

# --- Visualization ---

# Histogram of Magnitude of Error (ME)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
df['Magnitude of Error (ME)'].hist()
plt.title('Histogram of Magnitude of Error (ME)')
plt.xlabel('Magnitude of Error (Diopters)')
plt.ylabel('Frequency')

# Histogram of Angle of Error (AE)
plt.subplot(1, 2, 2)
df['Angle of Error (AE)'].hist(bins=30) # Adjust bins as needed
plt.title('Histogram of Angle of Error (AE)')
plt.xlabel('Angle of Error (Degrees)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

# Scatter plot of ME vs AE
plt.figure(figsize=(8, 6))
plt.scatter(df['Angle of Error (AE)'], df['Magnitude of Error (ME)'])
plt.title('Magnitude of Error (ME) vs. Angle of Error (AE)')
plt.xlabel('Angle of Error (Degrees)')
plt.ylabel('Magnitude of Error (Diopters)')
plt.grid(True)
plt.show()

# Polar plot showing ME and AE for individual patients
plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')

# Define colors for each patient (optional, or use a colormap based on something else if desired)
# For simplicity, using a basic color map for potentially more patients
cmap = plt.cm.get_cmap('hsv', len(df)) # Use HSV colormap for color diversity

for i in range(len(df)):
    me = df.iloc[i]['Magnitude of Error (ME)']
    ae_degrees = df.iloc[i]['Angle of Error (AE)']
    ae_radians = np.deg2rad(ae_degrees) # Convert AE to radians

    ax.plot([0, ae_radians], [0, me], color=cmap(i), marker='o', markersize=4, linestyle='-', linewidth=0.7, alpha=0.6)

ax.set_theta_zero_location("N") # Set 0 degrees to the top (North)
ax.set_theta_direction(-1)     # Set angle direction clockwise
ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
ax.set_title("Magnitude and Angle of Error (ME & AE) for Patients", va='bottom')

# Set r-limits dynamically based on the max ME
max_me = df['Magnitude of Error (ME)'].max() + 0.5 # Add some padding
ax.set_rlim(0, max_me)
ax.set_rticks(np.linspace(0, max_me, 5)) # Adjust radial ticks

ax.grid(True)

# Optional: Add a legend, but with many points, it might be cluttered.
# You could add a color bar if coloring by another variable.
# ax.legend(...)

plt.show()


def plot_target_induced_astigmatism_vectors(df, target_magnitude, target_axis_degrees):
    """
    Calculates and plots the SIA vectors relative to a target induced astigmatism vector.

    Args:
        df (pd.DataFrame): The DataFrame containing SIA data.
        target_magnitude (float): The target SIA magnitude.
        target_axis_degrees (float): The target SIA axis in degrees (0-180).
    """
    # Convert target vector to Cartesian components
    target_axis_radians = np.deg2rad(target_axis_degrees)
    target_X = target_magnitude * np.cos(target_axis_radians * 2)
    target_Y = target_magnitude * np.sin(target_axis_radians * 2)

    print(f"--- Target Induced Astigmatism ---")
    print(f"Target Magnitude: {target_magnitude:.2f}, Target Axis: {target_axis_degrees:.2f}°")
    print(f"Target X (Double Angle): {target_X:.2f}, Target Y (Double Angle): {target_Y:.2f}")
    print("----------------------------------")

    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot the target vector
    ax.plot([0, target_axis_radians], [0, target_magnitude], color='black', marker='*', markersize=10, linestyle='--', linewidth=2, label='Target SIA Vector')

    # Plot individual patient SIA vectors
    cmap = plt.cm.viridis # Colormap for individual vectors
    norm = plt.Normalize(df['SIA-vector-magnitude'].min(), df['SIA-vector-magnitude'].max())

    for i in range(len(df)):
        sia_magnitude = df.iloc[i]['SIA-vector-magnitude']
        sia_axis_degrees = df.iloc[i]['SIA-vector-axis']
        sia_axis_radians = np.deg2rad(sia_axis_degrees)

        color = cmap(norm(sia_magnitude))

        ax.plot([0, sia_axis_radians], [0, sia_magnitude], color=color, linestyle='-', linewidth=0.5, alpha=0.6) # Plot as lines

    ax.set_theta_zero_location("N") # Set 0 degrees to the top (North)
    ax.set_theta_direction(-1)     # Set angle direction clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title("SIA Vectors Relative to Target Vector (Colored by Magnitude)", va='bottom')

    # Set r-limits dynamically
    max_r = max(df['SIA-vector-magnitude'].max(), target_magnitude) + 1 # Add some padding
    ax.set_rlim(0, max_r)
    ax.set_rticks(np.linspace(0, max_r, 5)) # Adjust radial ticks

    # Add color bar for magnitude
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = plt.colorbar(sm, ax=ax, label='SIA Vector Magnitude')

    # Add grid
    ax.grid(True)

    ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5))

    plt.show()

# Example usage: Assuming a target SIA magnitude of 0.5 D at an axis of 90 degrees
target_magnitude_example = 0.5
target_axis_degrees_example = 90

plot_target_induced_astigmatism_vectors(df, target_magnitude_example, target_axis_degrees_example)


def plot_double_angle_vectors_with_labels(df, num_patients=5):
    """
    Plots pre-operative, post-operative, and SIA vectors for a specified number of patients
    in the double-angle vector space (polar plot) with labels.

    Args:
        df (pd.DataFrame): The DataFrame containing pre-op, post-op, and SIA vector data.
        num_patients (int): The number of patients to plot (default is 5).
    """
    plt.figure(figsize=(10, 10))
    ax = plt.subplot(111, projection='polar')

    # Get unique patient indices to plot
    patient_indices_to_plot = df.index[:min(num_patients, len(df))]

    # Define distinct colors for each patient
    # Using a larger colormap like 'tab20' or 'viridis' scaled to the number of patients
    cmap_patients = plt.cm.get_cmap('tab20', len(patient_indices_to_plot))

    # Define colors for vector types (pre-op, post-op, SIA)
    color_preop = 'blue'
    color_postop = 'green'
    color_sia = 'red'

    max_magnitude_overall = 0

    for i, patient_index in enumerate(patient_indices_to_plot):
        patient_data = df.loc[patient_index]

        # Pre-op Vector
        m1 = patient_data['m1']
        a1_degrees = patient_data['a1']
        a1_radians_double = np.deg2rad(a1_degrees * 2) # Double angle for plotting

        # Post-op Vector
        m2 = patient_data['m2']
        a2_degrees = patient_data['a2']
        a2_radians_double = np.deg2rad(a2_degrees * 2) # Double angle for plotting

        # SIA Vector
        sia_magnitude = patient_data['SIA-vector-magnitude']
        sia_axis_degrees = patient_data['SIA-vector-axis'] # Already 0-180
        sia_axis_radians_double = np.deg2rad(sia_axis_degrees * 2) # Double angle for plotting


        # Plot pre-op vector
        ax.plot([0, a1_radians_double], [0, m1], color=color_preop, linestyle='-', linewidth=1, alpha=0.7, label=f'Patient {patient_index} Pre-op' if i==0 else "")

        # Plot post-op vector
        ax.plot([0, a2_radians_double], [0, m2], color=color_postop, linestyle='-', linewidth=1, alpha=0.7, label=f'Patient {patient_index} Post-op' if i==0 else "")

        # Plot SIA vector (from the end of pre-op to the end of post-op) - This is the vector addition
        # Plotting the SIA vector from the origin as a separate arrow is also common.
        # Let's plot it from the origin in double-angle space for clarity
        ax.plot([0, sia_axis_radians_double], [0, sia_magnitude], color=color_sia, linestyle='--', linewidth=1.5, alpha=0.8, label=f'Patient {patient_index} SIA' if i==0 else "")


        # Update max magnitude for setting limits
        max_magnitude_overall = max(max_magnitude_overall, m1, m2, sia_magnitude)

    # Set properties for the polar plot
    ax.set_theta_zero_location("E") # Set 0 degrees to the right (often used in double angle plots)
    ax.set_theta_direction(1)     # Set angle direction counter-clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title(f"Double Angle Vectors for {min(num_patients, len(df))} Patients", va='bottom')

    # Set r-limits dynamically
    plot_limit = max_magnitude_overall + 0.5 # Add some padding
    ax.set_rlim(0, plot_limit)
    ax.set_rticks(np.linspace(0, plot_limit, 5)) # Adjust radial ticks

    # Add grid
    ax.grid(True)

    # Create custom legend handles for vector types
    preop_legend = plt.Line2D([0], [0], color=color_preop, lw=1, label='Pre-op Vector')
    postop_legend = plt.Line2D([0], [0], color=color_postop, lw=1, label='Post-op Vector')
    sia_legend = plt.Line2D([0], [0], color=color_sia, lw=1.5, linestyle='--', label='SIA Vector')

    # Combine the custom legend handles
    legend_handles = [preop_legend, postop_legend, sia_legend]

    # Add legend
    ax.legend(handles=legend_handles, loc='lower left', bbox_to_anchor=(1.05, 0.5)) # Place legend outside the plot


    plt.show()

# Example usage: Plot the double angle vectors for the first 5 patients
plot_double_angle_vectors_with_labels(df, num_patients=5)

# You can change the number of patients
# plot_double_angle_vectors_with_labels(df, num_patients=10)


def plot_vector_addition_in_double_angle_space(df, patient_index):
    """
    Illustrates the vector addition (pre-op + SIA = post-op) in the double-angle vector space
    for a single patient.

    Args:
        df (pd.DataFrame): The DataFrame containing pre-op, post-op, and SIA vector data.
        patient_index (int): The index of the patient in the DataFrame.
    """
    if patient_index < 0 or patient_index >= len(df):
        print(f"Error: Patient index {patient_index} is out of bounds.")
        return

    patient_data = df.iloc[patient_index]

    # Get vector components
    m1 = patient_data['m1']
    a1_degrees = patient_data['a1']
    a1_radians_double = np.deg2rad(a1_degrees * 2)

    m2 = patient_data['m2']
    a2_degrees = patient_data['a2']
    a2_radians_double = np.deg2rad(a2_degrees * 2)

    sia_magnitude = patient_data['SIA-vector-magnitude']
    sia_axis_degrees = patient_data['SIA-vector-axis']
    sia_axis_radians_double = np.deg2rad(sia_axis_degrees * 2)

    # Convert pre-op, post-op, and SIA to Cartesian coordinates in double-angle space
    x1 = patient_data['x1']
    y1 = patient_data['y1']

    x2 = patient_data['x2']
    y2 = patient_data['y2']

    xsia = patient_data['Xsia']
    ysia = patient_data['Ysia']


    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot Pre-op vector from origin
    ax.plot([0, a1_radians_double], [0, m1], color='blue', marker='o', markersize=5, linestyle='-', linewidth=2, label='Pre-op Vector')
    ax.annotate(f'Pre-op\nM={m1:.2f}\nA={a1_degrees:.0f}°',
                xy=(a1_radians_double, m1), xycoords='data',
                xytext=(5, 5), textcoords='offset points',
                fontsize=9, ha='left', va='bottom', color='blue')

    # Plot SIA vector starting from the end of the Pre-op vector (vector addition)
    # The end point of the pre-op vector is (x1, y1) in Cartesian
    # The SIA vector in Cartesian is (xsia, ysia)
    # The end point of the SIA vector added to pre-op is (x1 + xsia, y1 + ysia) which should equal (x2, y2)
    # In polar coordinates, we draw the SIA vector starting from the (x1, y1) location.
    # We need to convert (x1, y1) to polar coordinates to start drawing.
    # The SIA vector is (xsia, ysia) in Cartesian. Its magnitude is sia_magnitude and double angle is sia_axis_radians_double.
    # To plot from (x1, y1), we translate the origin.
    # The end point will be at (x1 + xsia, y1 + ysia).
    # We can draw an arrow from (x1, y1) to (x1+xsia, y1+ysia) in Cartesian and then convert to polar.
    # A simpler way in polar plot is to draw the SIA vector from (x1,y1) by defining the start and end points in Cartesian and using annotate or drawing lines.

    # Let's draw the SIA vector as a dotted line from the end of pre-op to the end of post-op
    # Start point of SIA vector in Cartesian is (x1, y1)
    # End point of SIA vector in Cartesian is (x2, y2)
    ax.plot([np.arctan2(y1, x1), np.arctan2(y2, x2)], [np.sqrt(x1**2 + y1**2), np.sqrt(x2**2 + y2**2)],
            color='red', linestyle=':', linewidth=2, alpha=0.8)

    # Alternatively, plot SIA from the origin in double-angle space
    ax.plot([0, sia_axis_radians_double], [0, sia_magnitude], color='red', marker='x', markersize=5, linestyle='--', linewidth=2, label='SIA Vector')
    ax.annotate(f'SIA\nM={sia_magnitude:.2f}\nA={sia_axis_degrees:.0f}°',
                xy=(sia_axis_radians_double, sia_magnitude), xycoords='data',
                xytext=(-5, -5), textcoords='offset points', ha='right', va='top',
                fontsize=9, color='red')


    # Plot Post-op vector from origin
    ax.plot([0, a2_radians_double], [0, m2], color='green', marker='^', markersize=5, linestyle='-', linewidth=2, label='Post-op Vector')
    ax.annotate(f'Post-op\nM={m2:.2f}\nA={a2_degrees:.0f}°',
                xy=(a2_radians_double, m2), xycoords='data',
                xytext=(5, 5), textcoords='offset points',
                fontsize=9, ha='left', va='bottom', color='green')


    # Set properties for the polar plot
    ax.set_theta_zero_location("E") # Set 0 degrees to the right (often used in double angle plots)
    ax.set_theta_direction(1)     # Set angle direction counter-clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title(f"Vector Addition: Pre-op + SIA = Post-op (Patient Index {patient_index})", va='bottom')

    # Set r-limits dynamically
    max_magnitude_overall = max(m1, m2, sia_magnitude)
    plot_limit = max_magnitude_overall + 0.5 # Add some padding
    ax.set_rlim(0, plot_limit)
    ax.set_rticks(np.linspace(0, plot_limit, 5)) # Adjust radial ticks

    # Add grid
    ax.grid(True)

    # Add legend
    ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5))

    plt.show()

# Example usage: Plot the vector addition for the first patient (index 0)
plot_vector_addition_in_double_angle_space(df, 0)

# You can change the index to visualize for a different patient
# plot_vector_addition_in_double_angle_space(df, 5)


def plot_surgically_induced_astigmatism_analysis(df, patient_index=None, num_patients_polar=5, target_magnitude=None, target_axis_degrees=None):
    """
    Provides a comprehensive analysis and visualization of SIA.

    Args:
        df (pd.DataFrame): The DataFrame containing calculated SIA data.
        patient_index (int, optional): Index of a single patient to focus on for detailed plots.
                                       If None, only summary plots are generated.
        num_patients_polar (int): Number of individual patient vectors to show in the polar plots
                                  if patient_index is None. Defaults to 5.
        target_magnitude (float, optional): Target SIA magnitude for target vector plot.
        target_axis_degrees (float, optional): Target SIA axis in degrees for target vector plot.
    """

    print("--- Surgically Induced Astigmatism (SIA) Analysis ---")

    # 1. Summary Statistics and Histograms
    print("\nSummary Statistics:")
    print(df[['SIA-vector-magnitude', 'SIA-vector-axis']].describe())

    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    df['SIA-vector-magnitude'].hist(bins=20)
    plt.title('Histogram of SIA Vector Magnitude')
    plt.xlabel('SIA Vector Magnitude (Diopters)')
    plt.ylabel('Frequency')

    plt.subplot(1, 2, 2)
    df['SIA-vector-axis'].hist(bins=40)
    plt.title('Histogram of SIA Vector Axis')
    plt.xlabel('SIA Vector Axis (Degrees)')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()

    # 2. Scatter plot of Magnitude vs Axis
    plt.figure(figsize=(8, 6))
    plt.scatter(df['SIA-vector-axis'], df['SIA-vector-magnitude'], alpha=0.6)
    plt.title('SIA Vector Magnitude vs. SIA Vector Axis')
    plt.xlabel('SIA Vector Axis (Degrees)')
    plt.ylabel('SIA Vector Magnitude (Diopters)')
    plt.grid(True)
    plt.show()

    # 3. Overall SIA Vector Distribution (Polar Plot)
    print("\nSIA Vector Distribution (Polar Plot):")
    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    cmap = plt.cm.viridis
    norm = plt.Normalize(df['SIA-vector-magnitude'].min(), df['SIA-vector-magnitude'].max())

    for i in range(len(df)):
        sia_magnitude = df.iloc[i]['SIA-vector-magnitude']
        sia_axis_degrees = df.iloc[i]['SIA-vector-axis']
        sia_axis_radians = np.deg2rad(sia_axis_degrees)

        color = cmap(norm(sia_magnitude))

        ax.plot([0, sia_axis_radians], [0, sia_magnitude], color=color, linestyle='-', linewidth=0.7, alpha=0.6)

    ax.set_theta_zero_location("N") # 0 degrees at top
    ax.set_theta_direction(-1)     # Clockwise
    ax.set_rlabel_position(-22.5)
    ax.set_title("Distribution of Individual SIA Vectors (Colored by Magnitude)", va='bottom')

    max_r = df['SIA-vector-magnitude'].max() + 0.5
    ax.set_rlim(0, max_r)
    ax.set_rticks(np.linspace(0, max_r, 5))

    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = plt.colorbar(sm, ax=ax, label='SIA Vector Magnitude')

    ax.grid(True)
    plt.show()

    # 4. Centroid SIA Vector
    centroid_Xsia = df['Xsia'].mean()
    centroid_Ysia = df['Ysia'].mean()
    centroid_magnitude = np.sqrt(centroid_Xsia**2 + centroid_Ysia**2)
    centroid_axis_radians = np.arctan2(centroid_Ysia, centroid_Xsia)
    centroid_axis_degrees = np.rad2deg(centroid_axis_radians)
    centroid_axis_degrees = centroid_axis_degrees + 180 if centroid_axis_degrees < 0 else centroid_axis_degrees

    print("\nCentroid SIA Vector for the Group:")
    print(f"  Magnitude: {centroid_magnitude:.2f} Diopters")
    print(f"  Axis: {centroid_axis_degrees:.2f} Degrees")

    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    # Plot the centroid vector
    ax.plot([0, np.deg2rad(centroid_axis_degrees)], [0, centroid_magnitude], color='red', marker='o', markersize=8, linestyle='-', linewidth=3, label='Centroid SIA Vector')

    ax.set_theta_zero_location("N") # 0 degrees at top
    ax.set_theta_direction(-1)     # Clockwise
    ax.set_rlabel_position(-22.5)
    ax.set_title("Centroid SIA Vector", va='bottom')

    max_r = max(df['SIA-vector-magnitude'].max(), centroid_magnitude) + 0.5
    ax.set_rlim(0, max_r)
    ax.set_rticks(np.linspace(0, max_r, 5))

    ax.grid(True)
    ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5))
    plt.show()


    # 5. Target Induced Astigmatism (if specified)
    if target_magnitude is not None and target_axis_degrees is not None:
         print("\nSIA Vectors Relative to Target Induced Astigmatism:")
         plot_target_induced_astigmatism_vectors(df, target_magnitude, target_axis_degrees)


    # 6. Individual Patient Analysis (if index is provided)
    if patient_index is not None:
        print(f"\n--- Detailed Analysis for Patient Index: {patient_index} ---")
        if patient_index < 0 or patient_index >= len(df):
            print(f"Error: Patient index {patient_index} is out of bounds.")
        else:
            patient_data = df.iloc[patient_index]
            print(f"Pre-op (M, A): ({patient_data['m1']:.2f}, {patient_data['a1']:.0f}°)")
            print(f"Post-op (M, A): ({patient_data['m2']:.2f}, {patient_data['a2']:.0f}°)")
            print(f"SIA Vector (Magnitude, Axis): ({patient_data['SIA-vector-magnitude']:.2f}, {patient_data['SIA-vector-axis']:.0f}°)")
            print(f"SIA Vector (X, Y): ({patient_data['Xsia']:.2f}, {patient_data['Ysia']:.2f})")

            # Plot individual SIA vector (Cartesian)
            plot_sia_vector_with_legend(df, patient_index)

            # Plot vector addition in double-angle space (Polar)
            plot_vector_addition_in_double_angle_space(df, patient_index)

    else: # If no specific patient index, plot individual vectors for a few patients in polar
         print(f"\nIndividual SIA Vectors for First {num_patients_polar} Patients (Polar Plot):")
         plot_polar_sia_vectors_with_legend(df, num_patients=num_patients_polar)


    print("\n--- Analysis Complete ---")


# Example Usage of the Comprehensive Function:

# To run the full analysis including summary plots and centroid:
plot_surgically_induced_astigmatism_analysis(df)

# To run the full analysis focusing on a specific patient (e.g., index 5):
# plot_surgically_induced_astigmatism_analysis(df, patient_index=5)

# To run the full analysis and plot relative to a target vector (e.g., 0.5 D @ 90 degrees):
# plot_surgically_induced_astigmatism_analysis(df, target_magnitude=0.5, target_axis_degrees=90)

# To run the full analysis focusing on a patient AND relative to a target vector:
# plot_surgically_induced_astigmatism_analysis(df, patient_index=5, target_magnitude=0.5, target_axis_degrees=90)


# Magnitude of Error (ME) is the same as SIA-vector-magnitude
# Plotting the distribution of ME
plt.figure(figsize=(8, 6))
df['Magnitude of Error (ME)'].hist(bins=20)
plt.title('Distribution of Magnitude of Error (ME)')
plt.xlabel('Magnitude of Error (Diopters)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()

# Plotting the distribution of Angle of Error (AE) - which is SIA-vector-axis
plt.figure(figsize=(8, 6))
df['Angle of Error (AE)'].hist(bins=30)
plt.title('Distribution of Angle of Error (AE)')
plt.xlabel('Angle of Error (Degrees)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()

# Limit the dataframe to the first 50 patients for plotting
df_50 = df.head(50).copy()

# Plot the distribution of SIA vectors for the first 50 patients (Polar Plot)
print("\nSIA Vector Distribution for the first 50 Patients (Polar Plot):")
plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')

cmap = plt.cm.viridis
norm = plt.Normalize(df_50['SIA-vector-magnitude'].min(), df_50['SIA-vector-magnitude'].max())

for i in range(len(df_50)):
    sia_magnitude = df_50.iloc[i]['SIA-vector-magnitude']
    sia_axis_degrees = df_50.iloc[i]['SIA-vector-axis']
    sia_axis_radians = np.deg2rad(sia_axis_degrees)

    color = cmap(norm(sia_magnitude))

    ax.plot([0, sia_axis_radians], [0, sia_magnitude], color=color, linestyle='-', linewidth=0.7, alpha=0.6)

ax.set_theta_zero_location("N") # 0 degrees at top
ax.set_theta_direction(-1)     # Clockwise
ax.set_rlabel_position(-22.5)
ax.set_title("Distribution of Individual SIA Vectors (First 50 Patients, Colored by Magnitude)", va='bottom')

max_r = df_50['SIA-vector-magnitude'].max() + 0.5
ax.set_rlim(0, max_r)
ax.set_rticks(np.linspace(0, max_r, 5))

sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax, label='SIA Vector Magnitude')

ax.grid(True)
plt.show()

# Calculate and plot the Centroid SIA Vector for the first 50 patients
centroid_Xsia_50 = df_50['Xsia'].mean()
centroid_Ysia_50 = df_50['Ysia'].mean()
centroid_magnitude_50 = np.sqrt(centroid_Xsia_50**2 + centroid_Ysia_50**2)
centroid_axis_radians_50 = np.arctan2(centroid_Ysia_50, centroid_Xsia_50)
centroid_axis_degrees_50 = np.rad2deg(centroid_axis_radians_50)
centroid_axis_degrees_50 = centroid_axis_degrees_50 + 180 if centroid_axis_degrees_50 < 0 else centroid_axis_degrees_50

print("\nCentroid SIA Vector for the First 50 Patients:")
print(f"  Magnitude: {centroid_magnitude_50:.2f} Diopters")
print(f"  Axis: {centroid_axis_degrees_50:.2f} Degrees")

plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')

# Plot the centroid vector for the first 50 patients
ax.plot([0, np.deg2rad(centroid_axis_degrees_50)], [0, centroid_magnitude_50], color='red', marker='o', markersize=8, linestyle='-', linewidth=3, label='Centroid SIA Vector (First 50)')

ax.set_theta_zero_location("N") # 0 degrees at top
ax.set_theta_direction(-1)     # Clockwise
ax.set_rlabel_position(-22.5)
ax.set_title("Centroid SIA Vector (First 50 Patients)", va='bottom')

max_r_50 = max(df_50['SIA-vector-magnitude'].max(), centroid_magnitude_50) + 0.5
ax.set_rlim(0, max_r_50)
ax.set_rticks(np.linspace(0, max_r_50, 5))

ax.grid(True)
ax.legend(loc='lower left', bbox_to_anchor=(1.05, 0.5))
plt.show()


# Plot Magnitude of Error (ME) vs Angle of Error (AE) for the first 50 patients
plt.figure(figsize=(8, 6))
plt.scatter(df_50['Angle of Error (AE)'], df_50['Magnitude of Error (ME)'], alpha=0.6)
plt.title('Magnitude of Error (ME) vs. Angle of Error (AE) (First 50 Patients)')
plt.xlabel('Angle of Error (Degrees)')
plt.ylabel('Magnitude of Error (Diopters)')
plt.grid(True)
plt.show()

# Plot ME and AE for individual patients in polar plot (First 50 patients)
plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')

cmap = plt.cm.get_cmap('hsv', len(df_50))

for i in range(len(df_50)):
    me = df_50.iloc[i]['Magnitude of Error (ME)']
    ae_degrees = df_50.iloc[i]['Angle of Error (AE)']
    ae_radians = np.deg2rad(ae_degrees)

    ax.plot([0, ae_radians], [0, me], color=cmap(i), marker='o', markersize=4, linestyle='-', linewidth=0.7, alpha=0.6)

ax.set_theta_zero_location("N") # 0 degrees at top
ax.set_theta_direction(-1)     # Clockwise
ax.set_rlabel_position(-22.5)
ax.set_title("Magnitude and Angle of Error (ME & AE) (First 50 Patients)", va='bottom')

max_me_50 = df_50['Magnitude of Error (ME)'].max() + 0.5
ax.set_rlim(0, max_me_50)
ax.set_rticks(np.linspace(0, max_me_50, 5))

ax.grid(True)
plt.show()


# Plot double angle vectors for the first 50 patients
def plot_double_angle_vectors_for_50_patients(df):
    """
    Plots pre-operative, post-operative, and SIA vectors for the first 50 patients
    in the double-angle vector space (polar plot).

    Args:
        df (pd.DataFrame): The DataFrame containing pre-op, post-op, and SIA vector data.
    """
    # Select the first 50 patients
    df_plot = df.head(50).copy()

    plt.figure(figsize=(10, 10))
    ax = plt.subplot(111, projection='polar')

    # Define colors for vector types (pre-op, post-op, SIA)
    color_preop = 'blue'
    color_postop = 'green'
    color_sia = 'red'

    max_magnitude_overall = 0

    for i in range(len(df_plot)):
        patient_data = df_plot.iloc[i]

        # Pre-op Vector
        m1 = patient_data['m1']
        a1_degrees = patient_data['a1']
        # Double angle for plotting in polar space
        a1_radians_double = np.deg2rad(a1_degrees * 2)

        # Post-op Vector
        m2 = patient_data['m2']
        a2_degrees = patient_data['a2']
        # Double angle for plotting in polar space
        a2_radians_double = np.deg2rad(a2_degrees * 2)

        # SIA Vector
        sia_magnitude = patient_data['SIA-vector-magnitude']
        sia_axis_degrees = patient_data['SIA-vector-axis'] # Already 0-180
        # Double angle for plotting in polar space
        sia_axis_radians_double = np.deg2rad(sia_axis_degrees * 2)


        # Plot pre-op vector
        ax.plot([0, a1_radians_double], [0, m1], color=color_preop, linestyle='-', linewidth=1, alpha=0.3)

        # Plot post-op vector
        ax.plot([0, a2_radians_double], [0, m2], color=color_postop, linestyle='-', linewidth=1, alpha=0.3)

        # Plot SIA vector from the origin in double-angle space
        ax.plot([0, sia_axis_radians_double], [0, sia_magnitude], color=color_sia, linestyle='--', linewidth=1.5, alpha=0.5)


        # Update max magnitude for setting limits
        max_magnitude_overall = max(max_magnitude_overall, m1, m2, sia_magnitude)

    # Set properties for the polar plot
    # Set 0 degrees to the right (often used in double angle plots)
    ax.set_theta_zero_location("E")
    # Set angle direction counter-clockwise
    ax.set_theta_direction(1)
    # Move radial labels away from plotted lines
    ax.set_rlabel_position(-22.5)
    ax.set_title("Double Angle Vectors for First 50 Patients", va='bottom')

    # Set r-limits dynamically
    plot_limit = max_magnitude_overall + 0.5 # Add some padding
    ax.set_rlim(0, plot_limit)
    # Adjust radial ticks
    ax.set_rticks(np.linspace(0, plot_limit, 5))

    # Add grid
    ax.grid(True)

    # Create custom legend handles for vector types
    preop_legend = plt.Line2D([0], [0], color=color_preop, lw=1, label='Pre-op Vector', alpha=0.7)
    postop_legend = plt.Line2D([0], [0], color=color_postop, lw=1, label='Post-op Vector', alpha=0.7)
    sia_legend = plt.Line2D([0], [0], color=color_sia, lw=1.5, linestyle='--', label='SIA Vector', alpha=0.8)

    # Combine the custom legend handles
    legend_handles = [preop_legend, postop_legend, sia_legend]

    # Add legend
    # Place legend outside the plot
    ax.legend(handles=legend_handles, loc='lower left', bbox_to_anchor=(1.05, 0.5))


    plt.show()

# Call the function to plot double angle vectors for the first 50 patients
plot_double_angle_vectors_for_50_patients(df)


def plot_vector_addition_for_50_patients_in_double_angle_space(df):
    """
    Illustrates the vector addition (pre-op + SIA = post-op) in the double-angle vector space
    for the first 50 patients.

    Args:
        df (pd.DataFrame): The DataFrame containing pre-op, post-op, and SIA vector data.
    """
    # Select the first 50 patients
    df_plot = df.head(50).copy()

    plt.figure(figsize=(10, 10))
    ax = plt.subplot(111, projection='polar')

    # Define colors for vector types (pre-op, post-op, SIA)
    color_preop = 'blue'
    color_postop = 'green'
    color_sia = 'red'

    max_magnitude_overall = 0

    for i in range(len(df_plot)):
        patient_data = df_plot.iloc[i]

        # Get vector components
        m1 = patient_data['m1']
        a1_degrees = patient_data['a1']
        a1_radians_double = np.deg2rad(a1_degrees * 2)

        m2 = patient_data['m2']
        a2_degrees = patient_data['a2']
        a2_radians_double = np.deg2rad(a2_degrees * 2)

        sia_magnitude = patient_data['SIA-vector-magnitude']
        sia_axis_degrees = patient_data['SIA-vector-axis']
        sia_axis_radians_double = np.deg2rad(sia_axis_degrees * 2)

        # Convert to Cartesian for calculating the path of the SIA vector
        x1 = patient_data['x1']
        y1 = patient_data['y1']
        x2 = patient_data['x2']
        y2 = patient_data['y2']


        # Plot Pre-op vector from origin
        ax.plot([0, a1_radians_double], [0, m1], color=color_preop, linestyle='-', linewidth=1, alpha=0.3)

        # Plot SIA vector as a dotted line from the end of pre-op to the end of post-op
        # Start point of SIA vector in Cartesian is (x1, y1)
        # End point of SIA vector in Cartesian is (x2, y2)
        # Convert start and end Cartesian points to polar for plotting the line
        start_r, start_theta = np.sqrt(x1**2 + y1**2), np.arctan2(y1, x1)
        end_r, end_theta = np.sqrt(x2**2 + y2**2), np.arctan2(y2, x2)

        # Need to handle wrapping around 0/360 degrees if the path crosses it
        # This is a simplification; for precise vector addition path, one would plot in Cartesian first.
        # However, for visualizing the connection in polar, we can just draw the line between the endpoints.
        ax.plot([start_theta, end_theta], [start_r, end_r], color=color_sia, linestyle=':', linewidth=1.5, alpha=0.5)


        # Plot Post-op vector from origin
        ax.plot([0, a2_radians_double], [0, m2], color=color_postop, linestyle='-', linewidth=1, alpha=0.3)


        # Update max magnitude for setting limits
        max_magnitude_overall = max(max_magnitude_overall, m1, m2, sia_magnitude)

    # Set properties for the polar plot
    ax.set_theta_zero_location("E") # Set 0 degrees to the right (often used in double angle plots)
    ax.set_theta_direction(1)     # Set angle direction counter-clockwise
    ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
    ax.set_title("Vector Addition: Pre-op + SIA = Post-op (First 50 Patients)", va='bottom')

    # Set r-limits dynamically
    plot_limit = max_magnitude_overall + 0.5 # Add some padding
    ax.set_rlim(0, plot_limit)
    ax.set_rticks(np.linspace(0, plot_limit, 5)) # Adjust radial ticks

    # Add grid
    ax.grid(True)

    # Create custom legend handles for vector types
    preop_legend = plt.Line2D([0], [0], color=color_preop, lw=1, label='Pre-op Vector', alpha=0.7)
    postop_legend = plt.Line2D([0], [0], color=color_postop, lw=1, label='Post-op Vector', alpha=0.7)
    # SIA vector path visualization
    sia_legend = plt.Line2D([0], [0], color=color_sia, lw=1.5, linestyle=':', label='SIA Vector (Path)', alpha=0.8)

    # Combine the custom legend handles
    legend_handles = [preop_legend, sia_legend, postop_legend] # Order as Pre-op -> SIA -> Post-op


    # Add legend
    ax.legend(handles=legend_handles, loc='lower left', bbox_to_anchor=(1.05, 0.5)) # Place legend outside the plot


    plt.show()

# Call the function to plot vector addition for the first 50 patients
plot_vector_addition_for_50_patients_in_double_angle_space(df)


# Assuming you have hypothetical target SIA values for each patient
# If you have actual target values, load them or calculate them here
# For demonstration, let's assume a target of 0 SIA for all patients (aiming for plano)
# Or, let's assume a target calculated based on pre-op astigmatism and intended correction

# For the purpose of demonstrating IAA and ISS, let's assume a simple scenario:
# The target is to induce SIA equal to the pre-operative astigmatism vector rotated 180 degrees
# (i.e., to correct the pre-op astigmatism). This is not always the clinical target,
# but it allows calculation of IAA/ISS.
# A more common target is 0 SIA for refractive procedures aiming for plano. Let's use that.

# Add hypothetical Target SIA Magnitude and Axis columns
# Let's assume the target is to induce ZERO astigmatism (aiming for plano)
# This means the target SIA vector is (0, 0)
df['Target SIA Magnitude'] = 0.0
df['Target SIA Axis'] = 0.0 # Axis doesn't matter if magnitude is 0

# If your target is to induce a specific amount and axis (e.g., for torics),
# you would load those values here.
# Example if aiming to correct the pre-op astigmatism:
# df['Target SIA Magnitude'] = df['m1']
# df['Target SIA Axis'] = df['a1'].apply(lambda x: x + 90 if x <= 90 else x - 90) # Roughly 90 degrees away from pre-op axis

# Calculate Assessed Induced Astigmatism (AIA)
# AIA is the SIA actually achieved (which you have already calculated)
df['Assessed Induced Astigmatism (AIA) Magnitude'] = df['SIA-vector-magnitude']
df['Assessed Induced Astigmatism (AIA) Axis'] = df['SIA-vector-axis']

# Calculate Index of Assessed Adjustment (IAA)
# IAA = AIA Magnitude / Target SIA Magnitude
# If Target SIA Magnitude is 0, IAA is undefined or handled as specific case.
# A common approach is to calculate the magnitude of the difference between
# AIA and Target SIA.
# Let's calculate the difference vector first.

# Convert Target SIA to Cartesian (double angle)
df['Target SIA X'] = df['Target SIA Magnitude'] * np.cos(np.deg2rad(df['Target SIA Axis'] * 2))
df['Target SIA Y'] = df['Target SIA Magnitude'] * np.sin(np.deg2rad(df['Target SIA Axis'] * 2))

# Calculate the difference vector (AIA - Target SIA)
df['Difference X'] = df['Xsia'] - df['Target SIA X']
df['Difference Y'] = df['Ysia'] - df['Target SIA Y']

# Calculate the Magnitude of the Difference Vector (Equivalent to Magnitude of Error from Target)
df['Magnitude of Difference (AIA - Target)'] = np.sqrt(df['Difference X']**2 + df['Difference Y']**2)

# IAA is often defined as the ratio of what was induced to what was intended to be induced.
# This is problematic when target magnitude is zero.
# A more robust approach related to adjustment is comparing AIA to Target SIA.
# Let's calculate IAA as originally intended: AIA / TIA
# Handle division by zero for Target SIA Magnitude = 0
df['Index of Assessed Adjustment (IAA)'] = df.apply(
    lambda row: row['Assessed Induced Astigmatism (AIA) Magnitude'] / row['Target SIA Magnitude'] if row['Target SIA Magnitude'] != 0 else np.nan, axis=1
)

# If target is 0, IAA is not typically calculated this way.
# In the case of Target SIA = 0, the 'Magnitude of Difference' already represents the error.
# If the goal is plano, IAA definition is less relevant than just the final residual astigmatism
# or the magnitude of SIA itself.

# Let's stick to the formula definition but acknowledge the zero target issue.
# IAA is the ratio of what was induced to what was intended to be induced.
# A value of 1 means the induced astigmatism matched the target magnitude.

# Calculate Index of Surgical Success (ISS)
# ISS = Post-op Magnitude / Pre-op Magnitude
# This assesses how much the surgery reduced the original astigmatism.
# ISS = 0 means all pre-op astigmatism was corrected. ISS = 1 means no change. ISS > 1 means increased astigmatism.
# Handle division by zero for Pre-op Magnitude = 0
df['Index of Surgical Success (ISS)'] = df.apply(
    lambda row: row['m2'] / row['m1'] if row['m1'] != 0 else np.nan, axis=1
)

print("\nCalculated IAA and ISS (assuming Target SIA Magnitude = 0):")
print(df[['Target SIA Magnitude', 'Assessed Induced Astigmatism (AIA) Magnitude', 'Index of Assessed Adjustment (IAA)',
          'm1', 'm2', 'Index of Surgical Success (ISS)']].head())

# Note: IAA will be NaN if Target SIA Magnitude is 0.
# If your clinical target is to achieve plano (Target SIA = 0), then focusing on
# the 'Magnitude of Difference (AIA - Target)' which equals the 'SIA-vector-magnitude'
# is more relevant than IAA. ISS is still relevant for assessing reduction of original astigmatism.

# --- Visualization of ISS ---
# Histograms of ISS
plt.figure(figsize=(8, 6))
df['Index of Surgical Success (ISS)'].hist(bins=20)
plt.title('Distribution of Index of Surgical Success (ISS)')
plt.xlabel('ISS (Post-op Mag / Pre-op Mag)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()

# Scatter plot of ISS vs Pre-op Magnitude
plt.figure(figsize=(8, 6))
plt.scatter(df['m1'], df['Index of Surgical Success (ISS)'], alpha=0.6)
plt.title('Index of Surgical Success (ISS) vs. Pre-op Magnitude')
plt.xlabel('Pre-op Magnitude (Diopters)')
plt.ylabel('Index of Surgical Success (ISS)')
plt.axhline(1, color='red', linestyle='--', linewidth=1, label='ISS = 1 (No Change)')
plt.ylim(0, df['Index of Surgical Success (ISS)'].max() + 0.5 if df['Index of Surgical Success (ISS)'].max() is not np.nan else 2) # Adjust ylim
plt.grid(True)
plt.legend()
plt.show()

# Visualization of Magnitude of Difference (AIA - Target) - when Target is 0, this is SIA Magnitude
if (df['Target SIA Magnitude'] == 0).all():
    print("\nMagnitude of Difference (AIA - Target) is equivalent to SIA Magnitude when Target SIA is 0.")
    # The SIA Magnitude plots already cover this.
else:
     plt.figure(figsize=(8, 6))
     df['Magnitude of Difference (AIA - Target)'].hist(bins=20)
     plt.title('Distribution of Magnitude of Difference (AIA - Target)')
     plt.xlabel('Magnitude of Difference (Diopters)')
     plt.ylabel('Frequency')
     plt.grid(True)
     plt.show()

     # Scatter plot of Magnitude of Difference vs Target SIA Magnitude
     plt.figure(figsize=(8, 6))
     plt.scatter(df['Target SIA Magnitude'], df['Magnitude of Difference (AIA - Target)'], alpha=0.6)
     plt.title('Magnitude of Difference vs. Target SIA Magnitude')
     plt.xlabel('Target SIA Magnitude (Diopters)')
     plt.ylabel('Magnitude of Difference (Diopters)')
     plt.grid(True)
     plt.show()




# Add columns for reporting
df['Report SIA Magnitude'] = df['SIA-vector-magnitude'].round(2)
df['Report SIA Axis'] = df['SIA-vector-axis'].round(0).astype(int) # Round axis to nearest degree

print("\nDataFrame with rounded SIA Magnitude and Axis for reporting:")
print(df[['Report SIA Magnitude', 'Report SIA Axis']].head())

# Optional: Export the updated dataframe with report columns to a new Excel file
# Make sure you have write access to the path in your Google Drive
# output_path = '/content/drive/MyDrive/Dr Aishwarya thesis excel sheet with SIA reports.xlsx'
# df.to_excel(output_path, index=False)
# print(f"\nDataFrame exported to: {output_path}")

```

```python
# prompt: prediction analysis on post op SIA

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
def plot_predicted_vs_actual_sia(df):
    """
    Plots Predicted SIA vs Actual SIA (Magnitude and Axis) if predicted SIA data is available.
    Assumes columns like 'Predicted SIA Magnitude' and 'Predicted SIA Axis' exist in the DataFrame.

    Args:
        df (pd.DataFrame): The DataFrame containing calculated Actual SIA and optionally Predicted SIA data.
    """
    if 'Predicted SIA Magnitude' not in df.columns or 'Predicted SIA Axis' not in df.columns:
        print("\nPredicted SIA columns not found in the DataFrame. Skipping prediction analysis plots.")
        return

    print("\n--- Predicted vs. Actual SIA Analysis ---")

    # Scatter plot: Actual SIA Magnitude vs. Predicted SIA Magnitude
    plt.figure(figsize=(8, 6))
    plt.scatter(df['Predicted SIA Magnitude'], df['SIA-vector-magnitude'], alpha=0.6)
    plt.title('Actual SIA Magnitude vs. Predicted SIA Magnitude')
    plt.xlabel('Predicted SIA Magnitude (Diopters)')
    plt.ylabel('Actual SIA Magnitude (Diopters)')
    # Add line of perfect prediction (y=x)
    max_mag = max(df['Predicted SIA Magnitude'].max(), df['SIA-vector-magnitude'].max())
    plt.plot([0, max_mag], [0, max_mag], color='red', linestyle='--', label='Perfect Prediction (y=x)')
    plt.grid(True)
    plt.legend()
    plt.show()

    # Scatter plot: Actual SIA Axis vs. Predicted SIA Axis
    # Note: Axis comparison can be tricky due to 0/180 wrap-around and double-angle space.
    # A simple scatter plot helps visualize general agreement but might not capture nuances.
    # Consider using circular statistics for more robust axis correlation.
    plt.figure(figsize=(8, 6))
    plt.scatter(df['Predicted SIA Axis'], df['SIA-vector-axis'], alpha=0.6)
    plt.title('Actual SIA Axis vs. Predicted SIA Axis')
    plt.xlabel('Predicted SIA Axis (Degrees, 0-180)')
    plt.ylabel('Actual SIA Axis (Degrees, 0-180)')
    plt.xlim(0, 180)
    plt.ylim(0, 180)
    # Add line of perfect prediction (y=x)
    plt.plot([0, 180], [0, 180], color='red', linestyle='--', label='Perfect Prediction (y=x)')
    plt.grid(True)
    plt.legend()
    plt.show()


    # Calculate Prediction Error (Vector Difference)
    # Convert Predicted SIA to Cartesian (double angle)
    df['Predicted SIA X'] = df['Predicted SIA Magnitude'] * np.cos(np.deg2rad(df['Predicted SIA Axis'] * 2))
    df['Predicted SIA Y'] = df['Predicted SIA Magnitude'] * np.sin(np.deg2rad(df['Predicted SIA Axis'] * 2))

    # Calculate the vector difference: Actual SIA - Predicted SIA
    df['Prediction Error X'] = df['Xsia'] - df['Predicted SIA X']
    df['Prediction Error Y'] = df['Ysia'] - df['Predicted SIA Y']

    # Calculate the Magnitude of Prediction Error
    df['Magnitude of Prediction Error'] = np.sqrt(df['Prediction Error X']**2 + df['Prediction Error Y']**2)

    # Calculate the Angle of Prediction Error
    df['Angle of Prediction Error (Radians)'] = np.arctan2(df['Prediction Error Y'], df['Prediction Error X'])
    df['Angle of Prediction Error (Degrees)'] = np.rad2deg(df['Angle of Prediction Error (Radians)'])

    # Adjust angle to be in a desired range (e.g., -90 to 90 or 0 to 180 for interpretation)
    # Using -90 to 90 might be more intuitive for error axis
    df['Angle of Prediction Error (Degrees Adjusted)'] = df['Angle of Prediction Error (Degrees)'].apply(
        lambda x: x + 180 if x < -90 else (x - 180 if x > 90 else x)
    )


    print("\nPrediction Error Statistics:")
    print(df[['Magnitude of Prediction Error', 'Angle of Prediction Error (Degrees Adjusted)']].describe())

    # Histogram of Magnitude of Prediction Error
    plt.figure(figsize=(8, 6))
    df['Magnitude of Prediction Error'].hist(bins=20)
    plt.title('Histogram of Magnitude of Prediction Error')
    plt.xlabel('Magnitude of Prediction Error (Diopters)')
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.show()

    # Histogram of Angle of Prediction Error
    plt.figure(figsize=(8, 6))
    df['Angle of Prediction Error (Degrees Adjusted)'].hist(bins=30)
    plt.title('Histogram of Angle of Prediction Error')
    plt.xlabel('Angle of Prediction Error (Degrees)')
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.show()


    # Polar plot of Prediction Error Vectors
    plt.figure(figsize=(8, 8))
    ax = plt.subplot(111, projection='polar')

    cmap_error = plt.cm.plasma # Colormap for error magnitude
    norm_error = plt.Normalize(df['Magnitude of Prediction Error'].min(), df['Magnitude of Prediction Error'].max())


    for i in range(len(df)):
        error_magnitude = df.iloc[i]['Magnitude of Prediction Error']
        error_axis_degrees = df.iloc[i]['Angle of Prediction Error (Degrees)'] # Use 0-360 range for polar plot
        error_axis_radians = np.deg2rad(error_axis_degrees)

        color = cmap_error(norm_error(error_magnitude))

        ax.plot([0, error_axis_radians], [0, error_magnitude], color=color, linestyle='-', linewidth=0.7, alpha=0.6)

    # Note: The interpretation of the axis in the error vector's polar plot can be tricky.
    # It represents the direction of the *discrepancy* between actual and predicted SIA.
    # Setting theta_zero_location and direction depends on desired interpretation (e.g., 0 at right, counter-clockwise).
    ax.set_theta_zero_location("E") # Set 0 degrees to the right
    ax.set_theta_direction(1)     # Set angle direction counter-clockwise
    ax.set_rlabel_position(-22.5)
    ax.set_title("Prediction Error Vectors (Actual SIA - Predicted SIA)", va='bottom')

    max_r_error = df['Magnitude of Prediction Error'].max() + 0.5
    ax.set_rlim(0, max_r_error)
    ax.set_rticks(np.linspace(0, max_r_error, 5))

    sm_error = plt.cm.ScalarMappable(cmap=cmap_error, norm=norm_error)
    sm_error.set_array([])
    cbar_error = plt.colorbar(sm_error, ax=ax, label='Magnitude of Prediction Error')


    ax.grid(True)
    plt.show()


    print("\n--- Prediction Analysis Complete ---")

# To run the prediction analysis, you need to add 'Predicted SIA Magnitude'
# and 'Predicted SIA Axis' columns to your DataFrame 'df'.
# For demonstration, let's add some hypothetical predicted data.
# In a real scenario, these would come from a prediction model.

# Add hypothetical predicted SIA data for demonstration
# This is just dummy data. Replace with your actual predicted values.
# Example: Predicting slightly lower SIA magnitude and slightly different axis
np.random.seed(42) # for reproducibility
if 'SIA-vector-magnitude' in df.columns and 'SIA-vector-axis' in df.columns:
    df['Predicted SIA Magnitude'] = df['SIA-vector-magnitude'] * (0.8 + 0.4 * np.random.rand(len(df))) # Example: 80-120% of actual magnitude
    df['Predicted SIA Axis'] = df['SIA-vector-axis'] + (np.random.rand(len(df)) - 0.5) * 20 # Example: +/- 10 degrees difference
    # Ensure predicted axis is within 0-180 range if necessary, or handle in vector conversion
    df['Predicted SIA Axis'] = df['Predicted SIA Axis'] % 180
    df.loc[df['Predicted SIA Axis'] < 0, 'Predicted SIA Axis'] += 180
else:
    print("\nSkipping hypothetical prediction data generation: SIA columns not found.")


# Call the prediction analysis function
plot_predicted_vs_actual_sia(df)
```

```python
# prompt: accuracy analysis

# Ensure necessary libraries are imported (already in the provided code)
# import pandas as pd
# import numpy as np
# import matplotlib.pyplot as plt

# --- Vector Calculations (from previous steps, ensure these columns exist in df) ---
# Assuming df already has 'm1', 'a1', 'm2', 'a2' (pre-op and post-op magnitude and axis)
# and 'Xsia', 'Ysia', 'SIA-vector-magnitude', 'SIA-vector-axis' are calculated.

# If not, calculate them:
# Function to convert polar to Cartesian (double angle)
def polar_to_cartesian_double_angle(magnitude, axis_degrees):
  # Convert axis to radians and double for double angle space
  axis_radians_double = np.deg2rad(axis_degrees * 2)
  x = magnitude * np.cos(axis_radians_double)
  y = magnitude * np.sin(axis_radians_double)
  return x, y

# Function to convert Cartesian (double angle) to polar
def cartesian_to_polar_clinical_axis(x, y):
  magnitude = np.sqrt(x**2 + y**2)
  angle_radians_double = np.arctan2(y, x)
  # Convert double angle radians back to clinical axis degrees (0-180)
  angle_degrees_double = np.rad2deg(angle_radians_double)
  # Adjust double angle to 0-360 range
  angle_degrees_double = angle_degrees_double % 360
  angle_degrees = angle_degrees_double / 2
  # Adjust clinical axis to 0-180 range
  angle_degrees = angle_degrees % 180
  return magnitude, angle_degrees

# Ensure these columns are calculated if they are missing
if 'x1' not in df.columns or 'y1' not in df.columns:
    df[['x1', 'y1']] = df.apply(lambda row: polar_to_cartesian_double_angle(row['m1'], row['a1']), axis=1, result_type='expand')

if 'x2' not in df.columns or 'y2' not in df.columns:
    df[['x2', 'y2']] = df.apply(lambda row: polar_to_cartesian_double_angle(row['m2'], row['a2']), axis=1, result_type='expand')

if 'Xsia' not in df.columns or 'Ysia' not in df.columns:
    # SIA vector in Cartesian is the difference: Post-op Vector - Pre-op Vector
    df['Xsia'] = df['x2'] - df['x1']
    df['Ysia'] = df['y2'] - df['y1']

if 'SIA-vector-magnitude' not in df.columns or 'SIA-vector-axis' not in df.columns:
    # Convert SIA Cartesian vector back to polar (magnitude and clinical axis)
    df[['SIA-vector-magnitude', 'SIA-vector-axis']] = df.apply(lambda row: cartesian_to_polar_clinical_axis(row['Xsia'], row['Ysia']), axis=1, result_type='expand')

# --- Accuracy Analysis ---

# Define Magnitude of Error (ME) and Angle of Error (AE)
# Based on the problem description, ME is SIA magnitude and AE is SIA axis.
df['Magnitude of Error (ME)'] = df['SIA-vector-magnitude']
df['Angle of Error (AE)'] = df['SIA-vector-axis']

print("\n--- Accuracy Analysis: Magnitude and Angle of Error ---")
print("Magnitude of Error (ME) and Angle of Error (AE) calculated.")
print(df[['Magnitude of Error (ME)', 'Angle of Error (AE)']].describe())


# --- Visualization of ME and AE ---

# Histogram of Magnitude of Error (ME)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
df['Magnitude of Error (ME)'].hist(bins=20)
plt.title('Histogram of Magnitude of Error (ME)')
plt.xlabel('Magnitude of Error (Diopters)')
plt.ylabel('Frequency')
plt.grid(True)

# Histogram of Angle of Error (AE)
plt.subplot(1, 2, 2)
df['Angle of Error (AE)'].hist(bins=30) # Adjust bins as needed
plt.title('Histogram of Angle of Error (AE)')
plt.xlabel('Angle of Error (Degrees, 0-180)')
plt.ylabel('Frequency')
plt.grid(True)

plt.tight_layout()
plt.show()

# Scatter plot of ME vs AE
plt.figure(figsize=(8, 6))
plt.scatter(df['Angle of Error (AE)'], df['Magnitude of Error (ME)'], alpha=0.6)
plt.title('Magnitude of Error (ME) vs. Angle of Error (AE)')
plt.xlabel('Angle of Error (Degrees, 0-180)')
plt.ylabel('Magnitude of Error (Diopters)')
plt.xlim(0, 180) # Limit x-axis to 0-180 for clinical axis
plt.grid(True)
plt.show()

# Polar plot showing ME and AE for individual patients
# This is effectively the same as the SIA Vector Distribution plot, but labeled with ME/AE terms
print("\nMagnitude and Angle of Error (ME & AE) Polar Plot:")
plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')

cmap = plt.cm.viridis # Colormap for magnitude
norm = plt.Normalize(df['Magnitude of Error (ME)'].min(), df['Magnitude of Error (ME)'].max())

for i in range(len(df)):
    me = df.iloc[i]['Magnitude of Error (ME)']
    ae_degrees = df.iloc[i]['Angle of Error (AE)']
    ae_radians = np.deg2rad(ae_degrees) # Convert AE to radians for polar plotting

    color = cmap(norm(me))

    # Plot vector from origin [0, angle_radians] with length magnitude
    ax.plot([0, ae_radians], [0, me], color=color, linestyle='-', linewidth=0.7, alpha=0.6)

# Set properties for the polar plot
ax.set_theta_zero_location("N") # Set 0 degrees to the top (North)
ax.set_theta_direction(-1)     # Set angle direction clockwise (clinical convention)
ax.set_rlabel_position(-22.5)  # Move radial labels away from plotted lines
ax.set_title("Magnitude and Angle of Error (ME & AE) for Patients (Colored by Magnitude)", va='bottom')

# Set r-limits dynamically based on the max ME
max_me = df['Magnitude of Error (ME)'].max() + 0.5 # Add some padding
ax.set_rlim(0, max_me)
ax.set_rticks(np.linspace(0, max_me, 5)) # Adjust radial ticks

# Add color bar for magnitude
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax, label='Magnitude of Error (ME)')

ax.grid(True)
plt.show()

print("\n--- Accuracy Analysis Complete ---")
```
