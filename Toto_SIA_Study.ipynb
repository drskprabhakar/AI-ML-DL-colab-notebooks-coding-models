```python
# Install openpyxl if not already installed
!pip install openpyxl

import pandas as pd

# --- Define the columns for the patient data ---
columns = [
    "Name",
    "Age (years)",
    "Gender (M/F)",
    "Eye (RE/LE)",
    "Pre-op Anterior Astigmatism (D)",
    "Pre-op Posterior Astigmatism (D)",
    "Pre-op Total Astigmatism (D)",
    "1-Wk Anterior Astigmatism (D)",
    "1-Wk Posterior Astigmatism (D)",
    "1-Wk Total Astigmatism (D)",
    "6-Wk Anterior Astigmatism (D)",
    "6-Wk Posterior Astigmatism (D)",
    "6-Wk Total Astigmatism (D)",
    "SIA at 1 Wk (Vector, D)",
    "SIA at 6 Wk (Vector, D)",
    "Change in SIA (1Wk → 6Wk)",
]

# --- Create an empty DataFrame ---
df = pd.DataFrame(columns=columns)

# --- Add a sample row (optional) ---
sample_row = {
    "Patient ID": "P001",
    "Age (years)": 65,
    "Sex (M/F)": "M",
    "Eye (RE/LE)": "RE",
    "Pre-op Anterior Astigmatism (D)": 1.25,
    "Pre-op Posterior Astigmatism (D)": 0.30,
    "Pre-op Total Astigmatism (D)": 1.55,
    "1-Week Anterior Astigmatism (D)": "",
    "1-Week Posterior Astigmatism (D)": "",
    "1-Week Total Astigmatism (D)": "",
    "6-Week Anterior Astigmatism (D)": "",
    "6-Week Posterior Astigmatism (D)": "",
    "6-Week Total Astigmatism (D)": "",
    "SIA at 1 Week (Vector, D)": "",
    "SIA at 6 Weeks (Vector, D)": "",
    "Change in SIA (1W → 6W)": "",
}
df = pd.concat([df, pd.DataFrame([sample_row])], ignore_index=True)

# --- Save to Excel ---
Total_SIA = "/content/Total SIA study.xlsx"
df.to_excel(Total_SIA, index=False)

print(f"Excel template created successfully: {Total_SIA}")


```
