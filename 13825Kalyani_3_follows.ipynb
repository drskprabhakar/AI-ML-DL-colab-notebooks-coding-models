## Compare and visualize

### Subtask:
Compare the analysis results of the three follow-up visits using appropriate plots.


**Reasoning**:
Create a bar plot to compare the distribution of refractive error classifications across the three dataframes.



```python
import matplotlib.pyplot as plt
import seaborn as sns

classification_day1 = day1AR['Classification'].value_counts()
classification_wk6 = wk6AR['Classification'].value_counts()
classification_mth3 = mth3AR['Classification'].value_counts()

classification_combined = pd.DataFrame({
    'Day 1': classification_day1,
    'Week 6': classification_wk6,
    'Month 3': classification_mth3
}).fillna(0)

classification_combined.plot(kind='bar', figsize=(12, 7))
plt.title('Distribution of Refractive Error Classifications Over Time')
plt.xlabel('Classification')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Reasoning**:
Combine the logMAR columns from the three dataframes into a single dataframe and create box plots to compare their distributions over time.



```python
logmar_data = pd.DataFrame({
    'Time Point': ['Day 1'] * len(day1AR) + ['Week 6'] * len(wk6AR) + ['Month 3'] * len(mth3AR),
    'logMAR': pd.concat([day1AR['day1 logMAR'], wk6AR['wk6 logMAR'], mth3AR['mth3 logMAR']])
})

plt.figure(figsize=(10, 6))
sns.boxplot(x='Time Point', y='logMAR', data=logmar_data)
plt.title('Distribution of logMAR Visual Acuity Over Time')
plt.xlabel('Time Point')
plt.ylabel('logMAR')
plt.show()
```

**Reasoning**:
Create histograms to visualize the distribution of 'AGE', 'SPH', and 'CYL' for each of the three dataframes to compare their distributions over time.



**Reasoning**:
The previous code block failed because the column name for SPH is different in the wk6AR dataframe ('SPH ') and mth3AR dataframe ('SPH.1'). Need to fix the column names before plotting the histograms.



```python
numerical_cols = ['AGE', 'SPH', 'CYL']

# Rename columns for consistency
wk6AR = wk6AR.rename(columns={'SPH ': 'SPH', 'CYL.1': 'CYL'})
mth3AR = mth3AR.rename(columns={'SPH.1': 'SPH', 'CYL ': 'CYL'})


for col in numerical_cols:
    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1)
    sns.histplot(day1AR[col], kde=True)
    plt.title(f'{col} Distribution - Day 1')

    plt.subplot(1, 3, 2)
    sns.histplot(wk6AR[col], kde=True)
    plt.title(f'{col} Distribution - Week 6')

    plt.subplot(1, 3, 3)
    sns.histplot(mth3AR[col], kde=True)
    plt.title(f'{col} Distribution - Month 3')

    plt.tight_layout()
    plt.show()
```
