# chi square test for age and sex

````python
# prompt: chi square test for 'Age', 'Sex'

import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

# Create a contingency table of 'Age' and 'Sex'
# For chi-square, we typically need categorical data.
# 'Age' here is continuous. We need to categorize it first.
# Let's categorize 'Age' into bins. You can adjust the bins as needed.
df['Age_Category'] = pd.cut(df['Age'], bins=[0, 18, 35, 60, np.inf], labels=['0-18', '19-35', '36-60', '60+'])

contingency_table = pd.crosstab(df['Age_Category'], df['Sex'])

# Perform the chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of freedom: {dof}")
print("Expected frequencies table:")
expected
````

````python
# prompt: generate scatter plot for pre-op KH', 'pre-op KV' and post-op KH', 'post-op KV',  with trend line and equation side by side

import pandas as pd
import matplotlib.pyplot as plt
def plot_scatter_with_trendline(df, x_col, y_col, title, ax):
  """
    Generates a scatter plot with a trend line and displays the equation.

    Args:
      df (pd.DataFrame): The input DataFrame.
      x_col (str): The column name for the x-axis.
      y_col (str): The column name for the y-axis.
      title (str): The title of the plot.
      ax (matplotlib.axes.Axes): The axes to plot on.
    """
  # Drop rows with missing values in the relevant columns
  df_cleaned = df[[x_col, y_col]].dropna()

  if not df_cleaned.empty:
    # Plot scatter plot
    ax.scatter(df_cleaned[x_col], df_cleaned[y_col])

    # Calculate and plot the trend line
    m, b, r_value, p_value, std_err = stats.linregress(df_cleaned[x_col], df_cleaned[y_col])
    ax.plot(df_cleaned[x_col], m * df_cleaned[x_col] + b, color='red', label=f'y = {m:.2f}x + {b:.2f}\nR² = {r_value**2:.2f}')

    ax.set_title(title)
    ax.set_xlabel(x_col)
    ax.set_ylabel(y_col)
    ax.legend()
  else:
    ax.text(0.5, 0.5, 'No data to plot', horizontalalignment='center', verticalalignment='center', transform=ax.transAxes)
    ax.set_title(title)
    ax.set_xlabel(x_col)
    ax.set_ylabel(y_col)


# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot for pre-op KH vs pre-op KV
plot_scatter_with_trendline(df, 'pre-op KH', 'pre-op KV', 'Scatter Plot of Pre-operative KV vs. Pre-operative KH', axes[0])

# Plot for post-op KH vs post-op KV
plot_scatter_with_trendline(df, 'post-op KH', 'post-op KV', 'Scatter Plot of Post-operative KV vs. Post-operative KH', axes[1])

plt.tight_layout()
plt.show()
````
