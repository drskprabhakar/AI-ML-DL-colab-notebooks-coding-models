```python
import numpy as np
import pandas as pd
import scipy.stats as stats

def age_stats_by_gender(df):
    results = []

    for gender, group in df.groupby('Gender'):
        age = group['Age'].dropna()
        n = len(age)
        count = age.count()


        mean = age.mean()
        sd = age.std()
        median = age.median()
        min_val = age.min()
        max_val = age.max()
        skew = age.skew()
        kurt = age.kurtosis()

        # 95% CI for mean
        se = sd / np.sqrt(n)
        ci_low, ci_high = stats.t.interval(0.95, df=n-1, loc=mean, scale=se)

        results.append([
            gender, count, mean, sd, median, min_val, max_val,
            skew, kurt, ci_low, ci_high
        ])

    # Table output
    columns = [
        'Gender', 'count', 'Mean', 'SD', 'Median', 'Min', 'Max',
        'Skewness', 'Kurtosis', '95% CI Lower', '95% CI Upper'
    ]

    return pd.DataFrame(results, columns=columns)

# Run the function
age_stats_gender_table = age_stats_by_gender(sorted_data)
print(age_stats_gender_table)
```

Perfect! We can classify astigmatism based on SteepK (steep corneal meridian) into WTR, ATR, or Oblique.
Here’s the common classification rule:
With-the-rule (WTR): SteepK between 60° and 120°
Against-the-rule (ATR): SteepK between 0°–30° or 150°–180°
Oblique: SteepK between 30°–60° or 120°–150°
Note: Angles are in degrees and modulo 180°.


```python
def classify_astigmatism(angle):
    if 60 <= angle <= 120:
        return 'WTR'
    elif (0 <= angle <= 30) or (150 <= angle <= 180):
        return 'ATR'
    else:  # 30-60 or 120-150
        return 'Oblique'

sorted_data['Preop_Astigmatism_Type'] = sorted_data['Preop_SteepK'].apply(classify_astigmatism)

sorted_data
```

```python
# Count of each astigmatism type
counts = sorted_data['Preop_Astigmatism_Type'].value_counts()

# Percentage of each type
percentages = sorted_data['Preop_Astigmatism_Type'].value_counts(normalize=True) * 100

# Combine into one table
astigmatism_summary = pd.DataFrame({
    'Count': counts,
    'Percentage (%)': percentages.round(2)
})

print(astigmatism_summary)
```

```python
import pandas as pd
import numpy as np
from scipy import stats

def descriptives(df, cols):
    results = {}

    for col in cols:
        data = df[col].dropna()
        n = data.count()
        mean = data.mean()
        sd = data.std()
        sem = stats.sem(data)  # standard error of mean
        ci = stats.t.interval(0.95, n-1, loc=mean, scale=sem)

        results[col] = {
            'Count': n,
            'Mean': mean,
            'SD': sd,
            'Median': data.median(),
            'Min': data.min(),
            'Max': data.max(),
            'Skewness': stats.skew(data),
            'Kurtosis': stats.kurtosis(data),
            '95% CI Lower': ci[0],
            '95% CI Upper': ci[1]
        }

    return pd.DataFrame(results).T

# Run on your dataset
descriptives(sorted_data, ['Preop_Cyl','Preop_MeanK'])
```

```python
import numpy as np


# Step 1: Convert positive cyl + flat-axis to negative-cylinder format
sorted_data['Preop_C_neg'] = -sorted_data['Preop_Cyl']
sorted_data['Preop_Axis_neg'] = (sorted_data['PreopA1'] + 90) % 180   # convert flat-meridian axis → steep-meridian axis

# Step 2: Calculate power vectors
axis_rad = np.deg2rad(sorted_data['Preop_Axis_neg'] * 2)

sorted_data['Preop_J0'] = -(sorted_data['Preop_C_neg'] / 2) * np.cos(axis_rad)
sorted_data['Preop_J45'] = -(sorted_data['Preop_C_neg'] / 2) * np.sin(axis_rad)
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# List of variables to summarize
cols = ['Preop_Cyl', 'Preop_MeanK', 'Preop_J0', 'Preop_J45']

summary_list = []

for col in cols:
    data = sorted_data[col].dropna()
    n = len(data)
    mean = data.mean()
    sd = data.std()
    median = data.median()
    minimum = data.min()
    maximum = data.max()
    skew = data.skew()
    kurtosis = data.kurtosis()

    # 95% CI for the mean
    se = sd / np.sqrt(n)
    ci_low, ci_high = stats.t.interval(0.95, df=n-1, loc=mean, scale=se)

    summary_list.append({
        'Variable': col,
        'Mean': mean,
        'SD': sd,
        'Median': median,
        'Min': minimum,
        'Max': maximum,
        'Skew': skew,
        'Kurtosis': kurtosis,
        '95% CI (Lower)': ci_low,
        '95% CI (Upper)': ci_high
    })

# Convert to table
summary_table = pd.DataFrame(summary_list)

summary_table
```
