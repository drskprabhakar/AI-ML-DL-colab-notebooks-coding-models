```python
import numpy as np
import pandas as pd

def get_steep_axis(flat_axis):
    """Convert flat meridian to steep meridian (0–180°)."""
    return (flat_axis + 90) % 180

def classify_astigmatism_from_steep(steep_axis):
    """Classify steep meridian into WTR, ATR, or Oblique."""
    if 60 <= steep_axis <= 120:
        return "WTR"
    elif (0 <= steep_axis <= 30) or (150 <= steep_axis <= 180):
        return "ATR"
    else:
        return "Oblique"

def preop_alpins_vectors(cyl, flat_axis):
    """Compute Alpins J0 and J45 vectors from cylinder and flat meridian."""
    steep_axis = get_steep_axis(flat_axis)
    rad = np.deg2rad(steep_axis * 2)  # double-angle formula
    J0 = -(cyl / 2) * np.cos(rad)
    J45 = -(cyl / 2) * np.sin(rad)
    return steep_axis, J0, J45

# Apply to DataFrame
def add_preop_vectors(df):
    # Compute steep axis
    df["SteepAxis_preop"] = df["PreopA1"].apply(get_steep_axis)

    # Classify astigmatism type
    df["AstigType_preop"] = df["SteepAxis_preop"].apply(classify_astigmatism_from_steep)

    # Compute Alpins vectors
    vectors = df.apply(lambda row: preop_alpins_vectors(row["PreopCyl"], row["PreopA1"]), axis=1)
    df[["SteepAxis_preop", "J0_preop", "J45_preop"]] = pd.DataFrame(vectors.tolist(), index=df.index)

    return df

# Example usage
sorteddata_no_outliers = add_preop_vectors(sorteddata_no_outliers)

# View first few rows
sorteddata_no_outliers[["PreopCyl", "PreopA1", "SteepAxis_preop", "AstigType_preop", "J0_preop", "J45_preop"]].head()
```

```python
import numpy as np

def classify_astigmatism_type(sph, cyl, threshold=0.1):
    """
    Classify astigmatism into Simple Myopic, Compound Myopic, Simple Hyperopic,
    Compound Hyperopic, and Mixed Astigmatism based on spherical and cylindrical power.
    """
    # Handle NaN values for Sph or Cyl
    if pd.isna(sph) or pd.isna(cyl):
        return np.nan

    # Convert 'Sph6wk ' to numeric if it's not already, coercing errors to NaN
    # This part was handled in a previous cell, but re-checking here for robustness
    # as the input to this function might be problematic if not converted earlier.
    # Assuming `sph` and `cyl` passed to the function are already numeric.

    M1 = sph
    M2 = sph + cyl

    # Simple Myopic Astigmatism
    if (abs(M1) < threshold and M2 < -threshold) or (abs(M2) < threshold and M1 < -threshold):
        return 'Simple Myopic Astigmatism'
    # Compound Hyperopic Astigmatism
    elif M1 > threshold and M2 > threshold:
        return 'Compound Hyperopic Astigmatism'
    # Compound Myopic Astigmatism
    elif M1 < -threshold and M2 < -threshold:
        return 'Compound Myopic Astigmatism'
    # Simple Hyperopic Astigmatism
    elif (abs(M1) < threshold and M2 > threshold) or (abs(M2) < threshold and M1 > threshold):
        return 'Simple Hyperopic Astigmatism'
    # Mixed Astigmatism
    elif (M1 < -threshold and M2 > threshold) or (M1 > threshold and M2 < -threshold):
        return 'Mixed Astigmatism'
    # No Astigmatism (both meridians close to zero or no significant cyl)
    elif abs(cyl) < threshold:
        return 'No Astigmatism'
    else:
        return np.nan # For any unclassified scenarios or if data is problematic

# Ensure 'Sph6wk ' and 'Cyl6wk' are numeric before applying the function
# 'Sph6wk ' was already handled in a previous step (cell 'vBhxn69dH3wj'), but ensure here too.
sorteddata_no_outliers['Sph6wk '] = pd.to_numeric(sorteddata_no_outliers['Sph6wk '], errors='coerce')
sorteddata_no_outliers['Cyl6wk'] = pd.to_numeric(sorteddata_no_outliers['Cyl6wk'], errors='coerce')

# Apply the function to create the new column
sorteddata_no_outliers['AstigmatismType_6wk'] = sorteddata_no_outliers.apply(
    lambda row: classify_astigmatism_type(row['Sph6wk '], row['Cyl6wk']),
    axis=1
)

# Display the first few rows with the new column
sorteddata_no_outliers[['Sph6wk ', 'Cyl6wk', 'AstigmatismType_6wk']].head()
```

```python
import numpy as np

def classify_astigmatism_type_3mth(sph, cyl, threshold=0.1):
    """
    Classify astigmatism into Simple Myopic, Compound Myopic, Simple Hyperopic,
    Compound Hyperopic, and Mixed Astigmatism based on spherical and cylindrical power.
    """
    # Handle NaN values for Sph or Cyl
    if pd.isna(sph) or pd.isna(cyl):
        return np.nan

    M1 = sph
    M2 = sph + cyl

    # Simple Myopic Astigmatism
    if (abs(M1) < threshold and M2 < -threshold) or (abs(M2) < threshold and M1 < -threshold):
        return 'Simple Myopic Astigmatism'
    # Compound Hyperopic Astigmatism
    elif M1 > threshold and M2 > threshold:
        return 'Compound Hyperopic Astigmatism'
    # Compound Myopic Astigmatism
    elif M1 < -threshold and M2 < -threshold:
        return 'Compound Myopic Astigmatism'
    # Simple Hyperopic Astigmatism
    elif (abs(M1) < threshold and M2 > threshold) or (abs(M2) < threshold and M1 > threshold):
        return 'Simple Hyperopic Astigmatism'
    # Mixed Astigmatism
    elif (M1 < -threshold and M2 > threshold) or (M1 > threshold and M2 < -threshold):
        return 'Mixed Astigmatism'
    # No Astigmatism (both meridians close to zero or no significant cyl)
    elif abs(cyl) < threshold:
        return 'No Astigmatism'
    else:
        return np.nan # For any unclassified scenarios or if data is problematic

# Ensure 'Sph3mth' and 'Cyl3mth' are numeric before applying the function
sorteddata_no_outliers['Sph3mth'] = pd.to_numeric(sorteddata_no_outliers['Sph3mth'], errors='coerce')
sorteddata_no_outliers['Cyl3mth'] = pd.to_numeric(sorteddata_no_outliers['Cyl3mth'], errors='coerce')

# Apply the function to create the new column
sorteddata_no_outliers['AstigmatismType_3mth'] = sorteddata_no_outliers.apply(
    lambda row: classify_astigmatism_type_3mth(row['Sph3mth'], row['Cyl3mth']),
    axis=1
)

# Display the first few rows with the new column
sorteddata_no_outliers[['Sph3mth', 'Cyl3mth', 'AstigmatismType_3mth']].head()
```

```python
import pandas as pd

# Combine the two Series into a DataFrame
# Using pd.concat with keys to create a MultiIndex for columns, then unstack to get desired format
contingency_table = pd.concat([astigmatism_counts, astigmatism_counts_3mth], axis=1, keys=['6 Weeks', '3 Months'])

# Fill any NaN values with 0
contingency_table = contingency_table.fillna(0).astype(int)

print("Contingency Table for Astigmatism Types at 6 Weeks and 3 Months:")
print(contingency_table)
```

```python
from scipy.stats import chi2_contingency

# Perform the Chi-square test for homogeneity
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square statistic: {chi2:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of freedom: {dof}")
print("Expected frequencies:\n", expected.round(2))

# Interpret the p-value
alpha = 0.05
if p_value < alpha:
    print(f"\nInterpretation: Since the p-value ({p_value:.4f}) is less than the significance level ({alpha}), we reject the null hypothesis.")
    print("This suggests that there is a statistically significant difference in the distribution of astigmatism types between 6 weeks and 3 months.")
else:
    print(f"\nInterpretation: Since the p-value ({p_value:.4f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.")
    print("This suggests that there is no statistically significant difference in the distribution of astigmatism types between 6 weeks and 3 months.")
```
