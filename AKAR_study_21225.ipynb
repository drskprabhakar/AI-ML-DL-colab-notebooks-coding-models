```python
def classify_astigmatism(axis):
    """
    Classify astigmatism based on steep meridian axis (0–180 degrees).
    """
    if (axis >= 1 and axis <= 30) or (axis >= 150 and axis <= 180):
        return "WTR"       # With-the-rule
    elif axis >= 60 and axis <= 120:
        return "ATR"       # Against-the-rule
    else:
        return "Oblique"

# Apply to your DataFrame column "Axis"
sorteddata_no_outliers["AstigType_preop"] = sorteddata_no_outliers["PreopA1"].apply(classify_astigmatism)

sorteddata_no_outliers[["PreopA1", "AstigType_preop"]].head()
```

```python
import scipy.stats as stats

# Observed counts
obs = sorteddata_no_outliers["AstigType_preop"].value_counts().sort_index()

# Expected counts = equal proportions
expected = [len(sorteddata_no_outliers) / 3] * 3

chi2, p = stats.chisquare(f_obs=obs, f_exp=expected)

# Calculate degrees of freedom separately
dof = len(obs) - 1
expected_freq = expected # already defined

print("Chi-square:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
print("Expected frequencies:", expected_freq)
print("\nConclusion:", "Significant difference" if p < 0.05 else "No significant difference")
```

```python
def get_steep_axis(PreopA1):
    """
    Convert flat meridian axis (0–180) to steep meridian.
    """
    steep = (PreopA1 + 90) % 180
    return steep


def classify_astigmatism_from_flat(PreopA1):
    """
    Classify WTR / ATR / Oblique using the steep meridian.
    """
    steep = get_steep_axis(PreopA1)

    if 60 <= steep <= 120:
        return "WTR"
    elif (0 <= steep <= 30) or (150 <= steep <= 180):
        return "ATR"
    else:
        return "Oblique"
```

```python
import numpy as np
import pandas as pd

def get_steep_axis(flat_axis):
    """Convert flat meridian to steep meridian (0–180°)."""
    return (flat_axis + 90) % 180

def classify_astigmatism_from_steep(steep_axis):
    """Classify steep meridian into WTR, ATR, or Oblique."""
    if 60 <= steep_axis <= 120:
        return "WTR"
    elif (0 <= steep_axis <= 30) or (150 <= steep_axis <= 180):
        return "ATR"
    else:
        return "Oblique"

def preop_alpins_vectors(cyl, flat_axis):
    """Compute Alpins J0 and J45 vectors from cylinder and flat meridian."""
    steep_axis = get_steep_axis(flat_axis)
    rad = np.deg2rad(steep_axis * 2)  # double-angle formula
    J0 = -(cyl / 2) * np.cos(rad)
    J45 = -(cyl / 2) * np.sin(rad)
    return steep_axis, J0, J45

# Apply to DataFrame
def add_preop_vectors(df):
    # Compute steep axis
    df["SteepAxis_preop"] = df["PreopA1"].apply(get_steep_axis)

    # Classify astigmatism type
    df["AstigType_preop"] = df["SteepAxis_preop"].apply(classify_astigmatism_from_steep)

    # Compute Alpins vectors
    vectors = df.apply(lambda row: preop_alpins_vectors(row["PreopCyl"], row["PreopA1"]), axis=1)
    df[["SteepAxis_preop", "J0_preop", "J45_preop"]] = pd.DataFrame(vectors.tolist(), index=df.index)

    return df

# Example usage
sorteddata_no_outliers = add_preop_vectors(sorteddata_no_outliers)

# View first few rows
sorteddata_no_outliers[["PreopCyl", "PreopA1", "SteepAxis_preop", "AstigType_preop", "J0_preop", "J45_preop"]].head()
```

```python
# Group by astigmatism type
summary = sorteddata_no_outliers.groupby("AstigType_preop").agg(
    Count=("PreopCyl", "count"),
    Percentage=("PreopCyl", lambda x: round(100 * len(x)/len(sorteddata_no_outliers), 2)),
    Mean_J0=("J0_preop", "mean"),
    Mean_J45=("J45_preop", "mean")
).reset_index()

print(summary)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Compute mean J0 and J45 for each astigmatism type
vector_means = sorteddata_no_outliers.groupby("AstigType_preop")[["J0_preop", "J45_preop"]].mean()

# Prepare polar coordinates
theta = np.arctan2(vector_means["J45_preop"], vector_means["J0_preop"])  # angle in radians
r = np.sqrt(vector_means["J0_preop"]**2 + vector_means["J45_preop"]**2)  # magnitude

# Plot
fig, ax = plt.subplots(subplot_kw={'projection':'polar'}, figsize=(7,7))

colors = {'WTR':'skyblue', 'ATR':'salmon', 'Oblique':'lightgreen'}

for i, astig_type in enumerate(vector_means.index):
    ax.arrow(theta[i], 0, 0, r[i], color=colors[astig_type],
             alpha=0.8, width=0.02, label=astig_type, edgecolor='black', linewidth=1.2)

# Formatting
ax.set_title("Preoperative Astigmatism - Mean Alpins Vectors", fontsize=14)
ax.set_rlabel_position(-22.5)  # move radial labels away
ax.set_rticks(np.arange(0, max(r)+0.1, 0.1))  # radial ticks
ax.set_theta_zero_location('E')  # 0 deg at east
ax.set_theta_direction(-1)       # clockwise

ax.legend(loc='upper right', fontsize=12)
plt.show()
```
