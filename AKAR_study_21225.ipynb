```python
import pandas as pd
import numpy as np
from scipy import stats

age = sorteddata['Age']

extended_stats = {
    'count': age.count(),
    'mean': age.mean(),
    'median': age.median(),
    'mode': age.mode()[0] if not age.mode().empty else None,
    'std_dev': age.std(),
    'variance': age.var(),
    'min': age.min(),
    'max': age.max(),
    'range': age.max() - age.min(),
    '25th_percentile': age.quantile(0.25),
    '50th_percentile': age.quantile(0.50),
    '75th_percentile': age.quantile(0.75),
    'IQR': age.quantile(0.75) - age.quantile(0.25),
    'skewness': age.skew(),
    'kurtosis': age.kurtosis()
}

pd.DataFrame.from_dict(extended_stats, orient='index', columns=['Age Statistics'])
```

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# Example: Replace this with your actual dataframe column
# age = df['Age'].dropna()
age = df['Age'].dropna()

# Compute mean and std for the normal curve
mu, sigma = age.mean(), age.std()

# Create figure with two plots side by side
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# ---------------------------
# 1. Histogram + Normal Curve
# ---------------------------
axs[0].hist(age, bins=15, density=True, alpha=0.6)
xmin, xmax = axs[0].get_xlim()

# X values for smooth normal curve
x = np.linspace(xmin, xmax, 200)
p = stats.norm.pdf(x, mu, sigma)

axs[0].plot(x, p, linewidth=2)
axs[0].set_title("Age Histogram with Normal (Bell) Curve")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Density")

# ---------------------------
# 2. QQ Plot
# ---------------------------
stats.probplot(age, dist="norm", plot=axs[1])
axs[1].set_title("QQ Plot for Age")

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import shapiro

# Drop missing values
age = df["Age"].dropna()

# Run Shapiro–Wilk test
stat, p = shapiro(age)

print("Shapiro–Wilk Test for Age")
print(f"Test Statistic : {stat:.4f}")
print(f"P-value        : {p:.4f}")

if p > 0.05:
    print("Interpretation : Age follows a normal distribution (fail to reject H0).")
else:
    print("Interpretation : Age does NOT follow a normal distribution (reject H0).")
```

```python
# Calculate Q1 (25th percentile) and Q3 (75th percentile) for 'PreopCyl'
Q1 = sorteddata['PreopCyl'].quantile(0.25)
Q3 = sorteddata['PreopCyl'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Define the upper and lower bounds for outlier detection
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Shape of DataFrame before outlier removal: {sorteddata.shape}")

# Filter out the outliers
sorteddata_no_outliers = sorteddata[(sorteddata['PreopCyl'] >= lower_bound) & (sorteddata['PreopCyl'] <= upper_bound)]

print(f"Shape of DataFrame after outlier removal: {sorteddata_no_outliers.shape}")

# You can now continue your analysis with 'sorteddata_no_outliers'
# or reassign it back to sorteddata if you prefer:
sorteddata = sorteddata_no_outliers
```
