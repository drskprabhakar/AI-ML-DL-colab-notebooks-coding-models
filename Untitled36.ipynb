```python
# prompt: Remove outliers in AL

# Assuming df['AL'] is a Series of numerical data
# Calculate Q1 (25th percentile)
Q1 = df['AL'].quantile(0.25)

# Calculate Q3 (75th percentile)
Q3 = df['AL'].quantile(0.75)

# Calculate IQR (Interquartile Range)
IQR = Q3 - Q1

# Define the lower and upper bounds to identify outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['AL'] < lower_bound) | (df['AL'] > upper_bound)]

# Remove outliers
df_cleaned = df[(df['AL'] >= lower_bound) & (df['AL'] <= upper_bound)]

print(f"Original shape: {df.shape}")
print(f"Shape after removing outliers: {df_cleaned.shape}")

# Optionally, visualize the data after removing outliers
df_cleaned[['AL']].boxplot()
```

```python
X=df[['KH','KV']]
y=df['AL']
```

```python
# prompt: Linear regression

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Create and train the Linear Regression model
regressor = LinearRegression()
regressor.fit(X_train, y_train)

# Make predictions on the test set
y_pred = regressor.predict(X_test)

# Evaluate the model
print('Mean Absolute Error:', metrics.mean_absolute_error(y_test, y_pred))
print('Mean Squared Error:', metrics.mean_squared_error(y_test, y_pred))
print('Root Mean Squared Error:', np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print('R-squared:', metrics.r2_score(y_test, y_pred))
```

```python
# prompt: Hyper parametric tuning

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
import xgboost as xgb
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Assuming X_train, y_train, X_test, y_test are defined from the cleaned data
# Re-defining them here for clarity in this response block
X = df_cleaned[['KH', 'KV']]
y = df_cleaned['AL']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)


# --- Random Forest Hyperparameter Tuning ---
print("\n--- Hyperparameter Tuning: Random Forest ---")
# Define the parameter grid
param_grid_rf = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Create a GridSearchCV object
grid_search_rf = GridSearchCV(estimator=RandomForestRegressor(random_state=0),
                              param_grid=param_grid_rf,
                              cv=5, # 5-fold cross-validation
                              scoring='r2', # Use R-squared as the scoring metric
                              n_jobs=-1) # Use all available cores

# Fit the grid search to the data
grid_search_rf.fit(X_train, y_train)

# Print the best parameters and best R-squared score
print("Best parameters for Random Forest:", grid_search_rf.best_params_)
print("Best cross-validation R-squared for Random Forest:", grid_search_rf.best_score_)

# Evaluate on the test set with the best estimator
best_rf_regressor = grid_search_rf.best_estimator_
y_pred_rf_tuned = best_rf_regressor.predict(X_test)
print('Tuned Random Forest R-squared on Test Set:', metrics.r2_score(y_test, y_pred_rf_tuned))


# --- XGBoost Hyperparameter Tuning ---
print("\n--- Hyperparameter Tuning: XGBoost ---")
# Define the parameter grid
param_grid_xgb = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 5, 7],
    'subsample': [0.7, 0.8, 0.9],
    'colsample_bytree': [0.7, 0.8, 0.9]
}

# Create a GridSearchCV object
grid_search_xgb = GridSearchCV(estimator=xgb.XGBRegressor(objective='reg:squarederror', random_state=0),
                               param_grid=param_grid_xgb,
                               cv=5,
                               scoring='r2',
                               n_jobs=-1)

# Fit the grid search to the data
grid_search_xgb.fit(X_train, y_train)

# Print the best parameters and best R-squared score
print("Best parameters for XGBoost:", grid_search_xgb.best_params_)
print("Best cross-validation R-squared for XGBoost:", grid_search_xgb.best_score_)

# Evaluate on the test set with the best estimator
best_xgb_regressor = grid_search_xgb.best_estimator_
y_pred_xgb_tuned = best_xgb_regressor.predict(X_test)
print('Tuned XGBoost R-squared on Test Set:', metrics.r2_score(y_test, y_pred_xgb_tuned))


# --- SVR Hyperparameter Tuning ---
print("\n--- Hyperparameter Tuning: SVR ---")
# It's highly recommended to scale features before SVR
# We'll use a pipeline for consistent scaling within cross-validation
pipeline_svr = Pipeline([
    ('scaler', StandardScaler()),
    ('svr', SVR())
])

# Define the parameter grid
param_grid_svr = {
    'svr__kernel': ['linear', 'rbf'], # Add 'poly', 'sigmoid' if relevant
    'svr__C': [0.1, 1, 10, 100],
    'svr__epsilon': [0.01, 0.1, 0.5]
}

# If using poly kernel, you might add:
# param_grid_svr['svr__degree'] = [2, 3, 4]

# Create a GridSearchCV object
grid_search_svr = GridSearchCV(estimator=pipeline_svr,
                               param_grid=param_grid_svr,
                               cv=5,
                               scoring='r2',
                               n_jobs=-1)

# Fit the grid search to the data
grid_search_svr.fit(X_train, y_train)

# Print the best parameters and best R-squared score
print("Best parameters for SVR:", grid_search_svr.best_params_)
print("Best cross-validation R-squared for SVR:", grid_search_svr.best_score_)

# Evaluate on the test set with the best estimator
best_svr_regressor = grid_search_svr.best_estimator_
y_pred_svr_tuned = best_svr_regressor.predict(X_test)
print('Tuned SVR R-squared on Test Set:', metrics.r2_score(y_test, y_pred_svr_tuned))


# --- K-Nearest Neighbors Hyperparameter Tuning ---
print("\n--- Hyperparameter Tuning: KNN ---")
# It's highly recommended to scale features before KNN
# We'll use a pipeline for consistent scaling within cross-validation
pipeline_knn = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', KNeighborsRegressor())
])

# Define the parameter grid
param_grid_knn = {
    'knn__n_neighbors': [3, 5, 7, 9, 11],
    'knn__weights': ['uniform', 'distance'],
    'knn__p': [1, 2] # 1 for Manhattan distance, 2 for Euclidean distance
}

# Create a GridSearchCV object
grid_search_knn = GridSearchCV(estimator=pipeline_knn,
                               param_grid=param_grid_knn,
                               cv=5,
                               scoring='r2',
                               n_jobs=-1)

# Fit the grid search to the data
grid_search_knn.fit(X_train, y_train)

# Print the best parameters and best R-squared score
print("Best parameters for KNN:", grid_search_knn.best_params_)
print("Best cross-validation R-squared for KNN:", grid_search_knn.best_score_)

# Evaluate on the test set with the best estimator
best_knn_regressor = grid_search_knn.best_estimator_
y_pred_knn_tuned = best_knn_regressor.predict(X_test)
print('Tuned KNN R-squared on Test Set:', metrics.r2_score(y_test, y_pred_knn_tuned))


# --- Decision Tree Hyperparameter Tuning ---
print("\n--- Hyperparameter Tuning: Decision Tree ---")
# Define the parameter grid
param_grid_dt = {
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'splitter': ['best', 'random']
}

# Create a GridSearchCV object
grid_search_dt = GridSearchCV(estimator=DecisionTreeRegressor(random_state=0),
                              param_grid=param_grid_dt,
                              cv=5,
                              scoring='r2',
                              n_jobs=-1)

# Fit the grid search to the data
grid_search_dt.fit(X_train, y_train)

# Print the best parameters and best R-squared score
print("Best parameters for Decision Tree:", grid_search_dt.best_params_)
print("Best cross-validation R-squared for Decision Tree:", grid_search_dt.best_score_)

# Evaluate on the test set with the best estimator
best_dt_regressor = grid_search_dt.best_estimator_
y_pred_dt_tuned = best_dt_regressor.predict(X_test)
print('Tuned Decision Tree R-squared on Test Set:', metrics.r2_score(y_test, y_pred_dt_tuned))
```
