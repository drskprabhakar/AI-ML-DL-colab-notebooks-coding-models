```python
# prompt: SIA magnitude and axis using double angle vector analysis

import matplotlib.pyplot as plt
import numpy as np
# Function to calculate SIA magnitude and axis using double angle vector analysis
def calculate_sia_double_angle(df):
    # Convert pre-op and post-op KH, KV to vectors
    # Assuming KH corresponds to the x-component and KV to the y-component for simplicity in this vector representation
    # Note: In ophthalmology, the relationship between K readings and vectors can be more complex,
    # but for a simplified vector representation using KH and KV differences, this approach is common.
    # The double angle vector analysis typically works with astigmatism values, not directly KH and KV.
    # To correctly apply double angle vector analysis (Javal's rule, Alpins method, etc.),
    # you would typically use the magnitude and axis of the astigmatism from pre-op and post-op.

    # This implementation will use the difference in KH and KV as components of a vector difference
    # and then apply a double angle transformation for the axis calculation, which is not the standard way
    # double angle vector analysis is applied to astigmatism, but follows the user's request to use KH and KV differences.
    # A more standard approach would require pre-op and post-op astigmatism values (cylinder and axis).

    # Calculate the vector difference components
    df['delta_KH'] = df['post-op KH'] - df['pre-op KH']
    df['delta_KV'] = df['post-op KV'] - df['pre-op KV']

    # Calculate the SIA magnitude using the difference in KH and KV
    # This is the same as the previous SIA vector magnitude calculation, essentially the hypotenuse
    # of a right triangle with sides delta_KH and delta_KV.
    df['SIA_magnitude_double_angle'] = np.sqrt(df['delta_KH']**2 + df['delta_KV']**2)

    # Calculate the SIA axis using double angle analysis
    # This part is where we attempt to adapt the double angle concept to delta_KH and delta_KV.
    # The standard double angle vector analysis works with astigmatism vectors represented as:
    # X = Cylinder * cos(2 * Axis)
    # Y = Cylinder * sin(2 * Axis)
    # And the resulting SIA vector (X_post - X_pre, Y_post - Y_pre) gives the SIA magnitude and axis.

    # Adapting this to delta_KH and delta_KV is not standard for calculating the SIA axis directly.
    # However, if we interpret delta_KH and delta_KV as the components of the SIA vector
    # in some transformed space (e.g., related to power vectors), we can find the angle in that space.

    # Let's calculate the angle of the vector (delta_KH, delta_KV) in a 2D Cartesian plane.
    # This angle is arctan(delta_KV / delta_KH).
    # To relate this to the surgical induced astigmatism axis, we need to consider the transformation to the double angle space.

    # If we assume delta_KH and delta_KV are proportional to the X and Y components of the SIA power vector in double angle space,
    # then the angle in the double angle space is atan2(delta_KV, delta_KH).
    # The corresponding axis is half of this angle.

    # Using atan2 for robustness with different quadrants
    double_angle_rad = np.arctan2(df['delta_KV'], df['delta_KH'])

    # Convert radians to degrees
    double_angle_deg = np.degrees(double_angle_rad)

    # Calculate the SIA axis (half of the double angle)
    sia_axis_double_angle = double_angle_deg / 2

    # Normalize the axis to be between 0 and 180 degrees
    # Handle negative angles
    sia_axis_double_angle = np.where(sia_axis_double_angle < 0, sia_axis_double_angle + 180, sia_axis_double_angle)
    # Handle angles greater than 180 (though typically double angle axis is 0-180, the result of / 2 could be larger temporarily)
    sia_axis_double_angle = np.where(sia_axis_double_angle > 180, sia_axis_double_angle - 180, sia_axis_double_angle)


    df['SIA_axis_double_angle'] = sia_axis_double_angle

    return df[['SIA_magnitude_double_angle', 'SIA_axis_double_angle']]

# Apply the function to the DataFrame
sia_results = calculate_sia_double_angle(df.copy()) # Use a copy to avoid modifying the original df if needed elsewhere

# Add the results back to the original DataFrame if you want
df['SIA_magnitude_double_angle'] = sia_results['SIA_magnitude_double_angle']
df['SIA_axis_double_angle'] = sia_results['SIA_axis_double_angle']

# Print the results
print("\nSIA Magnitude and Axis using Double Angle Vector Analysis (based on delta KH and delta KV):")
print(df[['SIA_magnitude_double_angle', 'SIA_axis_double_angle']].head())

# You can further analyze or visualize these results.
# For example, plot a vector diagram (not directly possible with just magnitude and axis columns, but conceptually you can plot).
# Or analyze the distribution of SIA magnitudes and axes.

# Example: Histogram of SIA Magnitude
plt.figure()
plt.hist(df['SIA_magnitude_double_angle'].dropna(), bins=20, edgecolor='black')
plt.xlabel('SIA Magnitude')
plt.ylabel('Frequency')
plt.title('Distribution of SIA Magnitude (Double Angle based on delta KH, delta KV)')
plt.show()

# Example: Polar plot of SIA Axis (Requires matplotlib's polar projection and converting axis to radians)
# Note: A proper polar plot of astigmatism vectors uses magnitude and double the axis.
# Here we are plotting the calculated SIA axis directly.
plt.figure()
ax = plt.subplot(111, projection='polar')
# Convert axis degrees to radians
axis_radians = np.radians(df['SIA_axis_double_angle'].dropna())
# Scatter plot on the polar axis
# You might want to use the SIA magnitude as the radial coordinate
ax.scatter(axis_radians, df['SIA_magnitude_double_angle'].dropna(), alpha=0.5)
ax.set_theta_zero_location("N") # Set 0 degrees to North (top)
ax.set_theta_direction(-1) # Set positive direction to clockwise
ax.set_title('SIA Axis Distribution (Double Angle based on delta KH, delta KV)', va='bottom')
plt.show()
```

```python
import math

def calculate_sia(pre_kh, pre_kv, pre_axis, post_kh, post_kv, post_axis):
    # Step 1: Compute pre-op and post-op astigmatism magnitudes
    A1 = pre_kv - pre_kh  # Pre-op magnitude
    A2 = post_kv - post_kh  # Post-op magnitude

    # Step 2: Convert to double-angle vector components
    theta1_rad = math.radians(2 * pre_axis)
    theta2_rad = math.radians(2 * post_axis)

    X1 = A1 * math.cos(theta1_rad)
    Y1 = A1 * math.sin(theta1_rad)

    X2 = A2 * math.cos(theta2_rad)
    Y2 = A2 * math.sin(theta2_rad)

    # Step 3: Calculate SIA vector components
    X_sia = X2 - X1
    Y_sia = Y2 - Y1

    # Step 4: Calculate SIA magnitude
    sia_magnitude = math.sqrt(X_sia**2 + Y_sia**2)

    # Step 5: Calculate SIA axis
    sia_axis_rad = 0.5 * math.atan2(Y_sia, X_sia)
    sia_axis_deg = math.degrees(sia_axis_rad)

    # Normalize angle to 0–180
    if sia_axis_deg < 0:
        sia_axis_deg += 180

    return sia_magnitude, sia_axis_deg

# Example input
pre_kh = 43.50
pre_kv = 44.00
pre_axis = 90

post_kh = 43.25
post_kv = 44.75
post_axis = 10

sia_mag, sia_axis = calculate_sia(pre_kh, pre_kv, pre_axis, post_kh, post_kv, post_axis)

print(f"SIA Magnitude: {sia_mag:.2f} D")
print(f"SIA Axis: {sia_axis:.2f}°")
```

```python
import math
import matplotlib.pyplot as plt

def calculate_sia_and_plot(pre_kh, pre_kv, pre_axis, post_kh, post_kv, post_axis):
    # Step 1: Calculate astigmatism magnitudes
    A1 = pre_kv - pre_kh
    A2 = post_kv - post_kh

    # Step 2: Convert axis to double-angle (in radians)
    theta1 = math.radians(2 * pre_axis)
    theta2 = math.radians(2 * post_axis)

    # Step 3: Convert to Cartesian components
    X1 = A1 * math.cos(theta1)
    Y1 = A1 * math.sin(theta1)

    X2 = A2 * math.cos(theta2)
    Y2 = A2 * math.sin(theta2)

    # Step 4: Calculate SIA vector
    X_sia = X2 - X1
    Y_sia = Y2 - Y1

    # Step 5: SIA magnitude and axis
    sia_magnitude = math.sqrt(X_sia**2 + Y_sia**2)
    sia_axis = math.degrees(0.5 * math.atan2(Y_sia, X_sia))
    if sia_axis < 0:
        sia_axis += 180

    # Step 6: Plotting
    plt.figure(figsize=(6, 6))
    plt.quiver(0, 0, X1, Y1, angles='xy', scale_units='xy', scale=1, color='blue', label='Pre-op')
    plt.quiver(0, 0, X2, Y2, angles='xy', scale_units='xy', scale=1, color='green', label='Post-op')
    plt.quiver(X1, Y1, X_sia, Y_sia, angles='xy', scale_units='xy', scale=1, color='red', label='SIA')

    # Format plot
    plt.xlim(-3, 3)
    plt.ylim(-3, 3)
    plt.axhline(0, color='gray', linewidth=0.5)
    plt.axvline(0, color='gray', linewidth=0.5)
    plt.gca().set_aspect('equal', adjustable='box')
    plt.title("SIA Vector Plot")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Print result
    print(f"SIA Magnitude: {sia_magnitude:.2f} D")
    print(f"SIA Axis: {sia_axis:.2f}°")

# Example
calculate_sia_and_plot(
    pre_kh=43.50,
    pre_kv=44.00,
    pre_axis=90,
    post_kh=43.25,
    post_kv=44.75,
    post_axis=10
)
```

```python
# prompt: SIA magnitude and axis describe

import matplotlib.pyplot as plt
import numpy as np
# The code provided already calculates SIA magnitude in two ways:
# 1. Using the difference in KH and KV directly as components (Pythagorean theorem): 'SIA_vector_magnitude'
#    df['SIA_vector_magnitude'] = np.sqrt(df['KH_diff']**2 + df['KV_diff']**2)
#    This calculates the magnitude of the vector representing the change from (pre-op KH, pre-op KV) to (post-op KH, post-op KV).
#    This isn't the standard ophthalmological SIA magnitude, which is based on astigmatism vectors.

# 2. Using a standard double-angle vector analysis function ('calculate_sia'): 'SIA_magnitude_calc' and 'SIA_axis_calc'
#    def calculate_sia(pre_kh, pre_kv, pre_axis, post_kh, post_kv, post_axis): ...
#    This function correctly computes the SIA magnitude and axis based on the change in astigmatism vectors,
#    derived from the pre-op and post-op K values and axes.
#    This method is the standard way to calculate Surgical Induced Astigmatism (SIA).

# The SIA magnitude and axis are added as new columns to the DataFrame:
# df['SIA_magnitude_calc']
# df['SIA_axis_calc']

# Let's describe what SIA magnitude and axis represent based on the standard calculation ('SIA_magnitude_calc' and 'SIA_axis_calc').

# SIA Magnitude ('SIA_magnitude_calc'):
# - Represents the amount of astigmatism induced by the surgery.
# - It is measured in Diopters (D).
# - It is the length of the vector difference between the post-operative astigmatism vector and the pre-operative astigmatism vector in double-angle space.
# - A higher SIA magnitude indicates a larger change in the astigmatism power.
# - It quantifies "how much" astigmatism was surgically induced or changed.

# SIA Axis ('SIA_axis_calc'):
# - Represents the orientation of the surgically induced astigmatism vector.
# - It is measured in degrees, typically between 0° and 180°.
# - It is half the angle of the SIA vector in the double-angle space.
# - It indicates the orientation of the cylinder power induced by the surgery. For example, an SIA axis of 90° suggests the induced astigmatism is oriented vertically, while 180° suggests it's horizontal.
# - It tells us "in which direction" the astigmatism was induced or changed.

# In summary, the SIA magnitude and axis provide a single vector representation of the total change in astigmatism caused by the surgical procedure, considering both the change in power and the change in orientation. The 'calculate_sia' function implements the standard method for deriving these values from pre- and post-operative keratometry readings and axes.

# The existing code already prints the first few rows of these calculated SIA values:
print("\nSIA Magnitude and Axis using Standard Calculation Method:")
print(df[['SIA_magnitude_calc', 'SIA_axis_calc']].head())

# You can perform further analysis on these columns, such as calculating summary statistics:
print("\nSummary Statistics for SIA Magnitude and Axis:")
print(df[['SIA_magnitude_calc', 'SIA_axis_calc']].describe())

# Or visualize their distributions as shown in the previous code for the double-angle based SIA (which is conceptually similar to the standard calculation):
# Histogram of SIA Magnitude
plt.figure()
plt.hist(df['SIA_magnitude_calc'].dropna(), bins=20, edgecolor='black')
plt.xlabel('SIA Magnitude (D)')
plt.ylabel('Frequency')
plt.title('Distribution of SIA Magnitude (Standard Calculation)')
plt.show()

# Polar plot of SIA Axis
plt.figure()
ax = plt.subplot(111, projection='polar')
axis_radians = np.radians(df['SIA_axis_calc'].dropna())
# Use SIA magnitude as the radial coordinate to show magnitude at each axis
ax.scatter(axis_radians, df['SIA_magnitude_calc'].dropna(), alpha=0.5)
ax.set_theta_zero_location("N") # Set 0 degrees to North (top)
ax.set_theta_direction(-1) # Set positive direction to clockwise
ax.set_title('SIA Axis Distribution (Standard Calculation)', va='bottom')
# Set theta ticks to represent axis angles
ax.set_xticks(np.radians(np.arange(0, 181, 30)))
ax.set_xticklabels([f'{i}°' for i in np.arange(0, 181, 30)])
plt.show()
```

```python
# prompt: Centroid plot

import matplotlib.pyplot as plt
import numpy as np
# Function to plot the centroid of the SIA vectors
def plot_sia_centroid(df):
  # Calculate the mean X and Y components of the SIA vectors
  # These X and Y components are already calculated implicitly within the calculate_sia function
  # Let's re-calculate them explicitly for the entire dataset
  X_sia_components = []
  Y_sia_components = []

  for index, row in df.iterrows():
    # Calculate the vector components for each row using the function logic
    A1 = row['pre-op KV'] - row['pre-op KH']
    A2 = row['post-op KV'] - row['post-op KH']
    theta1 = math.radians(2 * row['pre-op Axis'])
    theta2 = math.radians(2 * row['post-op Axis'])
    X1 = A1 * math.cos(theta1)
    Y1 = A1 * math.sin(theta1)
    X2 = A2 * math.cos(theta2)
    Y2 = A2 * math.sin(theta2)
    X_sia_components.append(X2 - X1)
    Y_sia_components.append(Y2 - Y1)

  df['SIA_X_component'] = X_sia_components
  df['SIA_Y_component'] = Y_sia_components

  # Calculate the mean SIA vector components (the centroid)
  mean_X_sia = df['SIA_X_component'].mean()
  mean_Y_sia = df['SIA_Y_component'].mean()

  # Calculate the magnitude and axis of the mean SIA vector
  centroid_magnitude = math.sqrt(mean_X_sia**2 + mean_Y_sia**2)
  centroid_axis_rad = 0.5 * math.atan2(mean_Y_sia, mean_X_sia)
  centroid_axis_deg = math.degrees(centroid_axis_rad)
  if centroid_axis_deg < 0:
      centroid_axis_deg += 180

  # Plotting
  plt.figure(figsize=(8, 8))
  ax = plt.subplot(111, projection='polar')

  # Plot individual SIA vectors (optional, can make the plot cluttered)
  # for x_sia, y_sia in zip(df['SIA_X_component'], df['SIA_Y_component']):
  #     ax.quiver(0, 0, 0.5 * math.atan2(y_sia, x_sia), math.sqrt(x_sia**2 + y_sia**2), angles='uv', scale_units='xy', scale=1, color='lightgray', alpha=0.3)

  # Plot the centroid vector
  # In a polar plot of astigmatism, the angle is 2*axis and radial is magnitude.
  # The centroid (mean X, mean Y) is already in double-angle space's Cartesian coordinates.
  # We can plot the centroid as a vector from the origin in this double-angle space.
  # Convert centroid Cartesian (X, Y) back to polar (Magnitude, 2*Axis)
  centroid_double_angle_rad = math.atan2(mean_Y_sia, mean_X_sia)
  # The radial coordinate is the centroid magnitude
  centroid_radial = centroid_magnitude # Using the magnitude of the mean vector

  # Plot the centroid vector on the polar plot
  ax.quiver(0, 0, centroid_double_angle_rad, centroid_radial, angles='uv', scale_units='xy', scale=1, color='red', label='Centroid SIA', width=0.005)


  ax.set_theta_zero_location("N") # Set 0 degrees to North (top)
  ax.set_theta_direction(-1) # Set positive direction to clockwise
  ax.set_title('SIA Centroid Plot', va='bottom')
  # Set theta ticks to represent axis angles (half of the double angle)
  ax.set_xticks(np.radians(np.arange(0, 360, 30))) # Double angle space is 0-360
  # Map the double angle tick labels to the corresponding axis angles (0-180)
  ax.set_xticklabels([f'{int(i/2)}°' for i in np.arange(0, 360, 30)])

  # Set radial limits based on the data magnitude range
  max_mag = df['SIA_magnitude_calc'].max()
  ax.set_rmax(max_mag * 1.2) # Extend the radial limit slightly

  plt.legend()
  plt.grid(True)
  plt.show()

  print(f"\nSIA Centroid Magnitude: {centroid_magnitude:.2f} D")
  print(f"SIA Centroid Axis: {centroid_axis_deg:.2f}°")

# Call the function to plot the centroid
plot_sia_centroid(df.copy()) # Use a copy to avoid modifying the original df if needed elsewhere
```
