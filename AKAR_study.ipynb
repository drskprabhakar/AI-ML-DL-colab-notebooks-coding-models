```python
from scipy.stats import shapiro

# Run Shapiro–Wilk test
stat, p = shapiro(sorteddata["PreopCyl"])

print("Shapiro–Wilk Test:")
print(f"Statistic = {stat:.6f}")
print(f"P-value   = {p:.6f}")

# Interpretation
alpha = 0.05
if p > alpha:
    print("Fail to reject H0 → Data looks normally distributed (p > 0.05)")
else:
    print("Reject H0 → Data is NOT normally distributed (p < 0.05)")
```

```python
def classify_astigmatism(axis):
    """
    Classify astigmatism based on steep meridian axis (0–180 degrees).
    """
    if (axis >= 1 and axis <= 15) or (axis >= 165 and axis <= 180):
        return "WTR"       # With-the-rule
    elif axis >= 75 and axis <= 105:
        return "ATR"       # Against-the-rule
    else:
        return "Oblique"

# Apply to your DataFrame column "Axis"
sorteddata["AstigType_preop"] = sorteddata["PreopA1"].apply(classify_astigmatism)

sorteddata[["PreopA1", "AstigType_preop"]].head()
```
