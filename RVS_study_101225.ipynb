PreopR1 = flatter meridian in diopters

PreopR2 = steeper meridian in diopters

PreopA1 = axis of the flatter meridian

This is standard for astigmatism classification.
The type of astigmatism is determined by the steep meridian, which is perpendicular to the flatter meridian:

Classification Rules (Using Steep Meridian)

Type	Steep meridian axis (degrees)

WTR	90 ± 30 → 60–120° (vertical steep meridian)

ATR	0 ± 30 → 0–30° or 150–180° (horizontal steep meridian)

Oblique	All others (31–59° and 121–149°)

Note: Since PreopA1 is flatter meridian, we add 90° and take modulo 180 to get the steep meridian axis.

```python
# Function to classify astigmatism
def classify_astig(r1_axis):
    steep_axis = (r1_axis + 90) % 180
    if 60 <= steep_axis <= 120:
        return 'WTR'
    elif (0 <= steep_axis <= 30) or (150 <= steep_axis <= 180):
        return 'ATR'
    else:
        return 'Oblique'

# Apply classification
sorted_data['Preop_Astig_Type'] = sorted_data['PreopA1'].apply(classify_astig)

# Count each type
type_counts = sorted_data['Preop_Astig_Type'].value_counts()
type_counts
```

```python
import pandas as pd
import numpy as np

# Calculate astigmatism magnitude
sorted_data['Preop_Astig_Mag'] = sorted_data['PreopR2'] - sorted_data['PreopR1']

# Calculate steep meridian axis
sorted_data['Steep_Axis'] = (sorted_data['PreopA1'] + 90) % 180

# Function to classify astigmatism based on steep meridian
def classify_astig(steep_axis):
    if 60 <= steep_axis <= 120:
        return 'WTR'
    elif (0 <= steep_axis <= 30) or (150 <= steep_axis <= 180):
        return 'ATR'
    else:
        return 'Oblique'

# Apply classification
sorted_data['Preop_Astig_Type'] = sorted_data['Steep_Axis'].apply(classify_astig)

# Select relevant columns for the table
astig_table = sorted_data[['PreopR1', 'PreopR2', 'PreopA1',
                           'Preop_Astig_Mag', 'Steep_Axis', 'Preop_Astig_Type']]

# Show table
astig_table.head(10)  # display first 10 rows
```
