```python
df['preop_kvkhdiff'] = df['pre-op KV']-df['pre-op KH']
```

```python
df['postop_kvkhdiff'] = df['post-op KV']-df['post-op KH']
```

```python
df['M1']= df['preop_kvkhdiff']
df['M2'] = df['postop_kvkhdiff']
df['A1'] = df['pre-op Axis']
df['A2'] = df['post-op Axis']
```

```python
df
```

```python
# find x and y
df['x1']= df['M1'] * np.cos(df['A1'])*2
df['y1'] = df['M1'] * np.sin(df['A1'])*2
df['x2'] = df['M2'] * np.cos(df['A2'])*2
df['y2'] = df['M2'] * np.sin(df['A2'])
```

```python
df
```

```python
# find Xsia and Ysia
df['Xsia'] = df['x2']-df['x1']
df['Ysia'] = df['y2']-df['y1']
```

```python
df
```

```python
# now apply pythogorean concept

df['sia_magnitude'] = np.sqrt(df['Xsia']**2 + df['Ysia']**2)
```

```python
df
```

```python
# now 0.5*tan inverse(ysia/xsia)

df['sia_axis'] = 0.5*np.arctan(df['Ysia']/df['Xsia'])*2
```

```python
df
```

```python
# convert radtodegree

df['sia_axisdegrees'] = np.rad2deg(df['sia_axis'])
```

```python
df
```

```python
# prompt: Quadrant handling

# Handling Quadrants for SIA axis
def adjust_sia_axis(row):
  x = row['Xsia']
  y = row['Ysia']
  axis = row['sia_axisdegrees']

  if x > 0 and y > 0: # Quadrant 1
    return axis
  elif x < 0 and y > 0: # Quadrant 2
    return axis + 180
  elif x < 0 and y < 0: # Quadrant 3
    return axis + 180
  elif x > 0 and y < 0: # Quadrant 4
    return axis + 360 if axis < 0 else axis # Ensure positive angle in Q4
  elif x == 0:
    if y > 0:
      return 90
    elif y < 0:
      return 270
    else:
      return 0 # Or NaN, depending on desired behavior for (0,0)
  elif y == 0:
    if x > 0:
      return 0
    elif x < 0:
      return 180
    else:
      return 0 # Or NaN

df['sia_axis_adjusted'] = df.apply(adjust_sia_axis, axis=1)

# Ensure angles are within 0-180 range after quadrant adjustment
df['sia_axis_adjusted'] = df['sia_axis_adjusted'] % 180

df
```

```python
# prompt: Plot preop, postop and sia vector

import matplotlib.pyplot as plt
import numpy as np
# Function to calculate X and Y components given magnitude and axis in degrees
def calculate_xy(magnitude, axis_degrees):
    # Convert axis from degrees to radians
    axis_rad = np.deg2rad(axis_degrees)
    # Calculate X and Y components (Note: Adjusting multiplication factor if needed, currently using 2 as in original code)
    x = magnitude * np.cos(axis_rad) * 2
    y = magnitude * np.sin(axis_rad) * 2 # Original code used * 2 for pre-op but * 1 for post-op y, ensure consistency based on definition
    return x, y

# Add columns for Pre-op X and Y
df['preop_x'], df['preop_y'] = zip(*df.apply(lambda row: calculate_xy(row['M1'], row['A1']), axis=1))

# Add columns for Post-op X and Y
df['postop_x'], df['postop_y'] = zip(*df.apply(lambda row: calculate_xy(row['M2'], row['A2']), axis=1))

# Plotting the vectors
plt.figure(figsize=(10, 10))
ax = plt.subplot(111)

# Plot preop vectors
for i in range(len(df)):
    ax.arrow(0, 0, df['preop_x'].iloc[i], df['preop_y'].iloc[i], head_width=0.2, head_length=0.3, fc='blue', ec='blue', label='Pre-op' if i == 0 else "", alpha=0.5)

# Plot postop vectors
for i in range(len(df)):
    ax.arrow(0, 0, df['postop_x'].iloc[i], df['postop_y'].iloc[i], head_width=0.2, head_length=0.3, fc='green', ec='green', label='Post-op' if i == 0 else "", alpha=0.5)

# Plot SIA vectors (which you've already calculated)
for i in range(len(df)):
     ax.arrow(df['preop_x'].iloc[i], df['preop_y'].iloc[i], df['Xsia'].iloc[i], df['Ysia'].iloc[i], head_width=0.2, head_length=0.3, fc='red', ec='red', label='SIA' if i == 0 else "", alpha=0.5)


# Set plot limits based on data range
max_coord = max(df[['preop_x', 'preop_y', 'postop_x', 'postop_y', 'Xsia', 'Ysia']].abs().max().max(), 1) * 1.2
ax.set_xlim([-max_coord, max_coord])
ax.set_ylim([-max_coord, max_coord])

# Add lines at x=0 and y=0
ax.axhline(0, color='grey', lw=0.8)
ax.axvline(0, color='grey', lw=0.8)

# Set aspect ratio to be equal to ensure vectors are not distorted
ax.set_aspect('equal', adjustable='box')

plt.title('Pre-op, Post-op, and SIA Vectors')
plt.xlabel('X Component')
plt.ylabel('Y Component')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# prompt: Remove outliers sia magnitude

import matplotlib.pyplot as plt
# Remove outliers for 'sia_magnitude' using the Interquartile Range (IQR) method
Q1 = df['sia_magnitude'].quantile(0.25)
Q3 = df['sia_magnitude'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out the outliers
df_filtered = df[(df['sia_magnitude'] >= lower_bound) & (df['sia_magnitude'] <= upper_bound)].copy()

print(f"Original DataFrame shape: {df.shape}")
print(f"Filtered DataFrame shape (after removing SIA magnitude outliers): {df_filtered.shape}")

# You can now continue your analysis and plotting using df_filtered
# For example, plot the boxplot of the filtered data to confirm outlier removal
plt.figure(figsize=(8, 6))
sns.boxplot(y=df_filtered['sia_magnitude'])
plt.title('Boxplot of SIA Magnitude (Outliers Removed)')
plt.ylabel('SIA Magnitude')
plt.show()

# You might also want to check the impact on other variables or replot other graphs using df_filtered
```
